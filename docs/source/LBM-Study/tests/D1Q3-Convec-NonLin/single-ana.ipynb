```python
%pylab inline
Nx=2501
c=4
plt = pyplot
raw = open(f'Densidade-{Nx}-c{c}').readlines()[:]
r1 = [ i.strip() for i in raw ]
r2 = ' '.join(r1)
import re
uv = [ float(i) for i in re.split(' * ', r2) ]
print(len(uv))
rhoA = np.array(uv[:int(len(uv)/2)])
rhoB = np.array(uv[int(len(uv)/2):])
```

```python
from scipy.optimize import brentq
sc=0.7071
ssx=linspace(0.0,20.0,Nx)*0
#print(ssx)
dx=20.0/(Nx-1)
def equation(ss):
        return 0.0+dx*i - (2.0 * ss * (1.0 - ss) / (ss**2 + (1.0 - ss)**2)**2.0) * 10.0
interval_start = 0.5
interval_end = 1.0
for i in range(0,len(ssx)):
    ssx[i] = brentq(equation, interval_start, interval_end)
dfsdx=2.0*ssx*(1.0-ssx)/(ssx**2+(1.0-ssx)**2)**2.0;
s=sc
f=s**2/(s**2+(1.0-s)**2)
ska=ssx*0
for i in range(0,len(ssx)):
    if (dfsdx[i]*10<10*f/sc and ssx[i]>sc):
    #if (dfsdx[i]*10<=10*f/sc and ssx[i]>=sc):
        ska[i]=ssx[i]
```
