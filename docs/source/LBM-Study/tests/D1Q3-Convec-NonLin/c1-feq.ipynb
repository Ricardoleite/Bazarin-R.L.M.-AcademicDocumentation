```python
for j in range(0,len(c0)):
    raw = open(f'Densidade-{Nx[0]}-c{c0[j]}').readlines()[:]
    r1 = [ i.strip() for i in raw ]
    r2 = ' '.join(r1)
    import re
    uv = [ float(i) for i in re.split(' * ', r2) ]
    #print(len(uv))
    rhoA = np.array(uv[:int(len(uv)/2)])
    rhoB = np.array(uv[int(len(uv)/2):])
    #=======================================================================================
    sc=0.7071
    ssx=linspace(0.0,20.0,Nx[0])*0
    dx=20.0/(Nx[0]-1)
    def equation(ss):
        return 0.0+dx*i - (2.0 * ss * (1.0 - ss) / (ss**2 + 1.0*(1.0 - ss)**2)**2.0) * 10.0
    interval_start = 0.5
    interval_end = 1.0
    for i in range(0,len(ssx)):
        ssx[i] = brentq(equation, interval_start, interval_end)
    dfsdx=2.0*ssx*(1.0-ssx)/(ssx**2+1.0*(1.0-ssx)**2)**2.0;
    s=sc
    f=s**2/(s**2+1.0*(1.0-s)**2)
    for i in range(0,len(ssx)):
        if (dfsdx[i]*10<10*f/sc and ssx[i]>sc):
            ska0[j][i]=ssx[i]
    #=======================================================================================
    L10[j]=sum(sqrt((ska0[j][:]-rhoA)**2))/Nx[0]
    print(f'L1-c{c0[j]}=',L10[j])
    L000[j]=sqrt(sum((ska0[j][:]-rhoA)**2))/Nx[0]
    print(f'L°°-c{c0[j]}=',L000[j])
```
