# Energy Dependent Discretization

Yes. You can extend the lecture’s formulation by introducing a **thermal energy per unit mass**, then deriving the discrete weights from its Gaussian moments. However, **changing the thermal velocity scale does not automatically change the weights**: that depends on whether you keep the discrete velocities fixed.

The lecture already introduces the reference velocity scale $(\sqrt{k_B T_0/m})$ and derives the lattice through Hermite orthogonality and prescribed abscissas, particularly in Sections 1.3 and 3 of Lecture06_A_velocity_discretization-3.pdf.

Below, I interpret your reference mass $(m_0)$ as a **molecular mass**, rather than a reference density or the mass represented by a lattice node.

**1. Introduce the energy and velocity scales**

For an ideal gas, define

$$
E_0=k_B T_0, \qquad \Theta_0=\frac{E_0}{m_0} =\frac{k_B T_0}{m_0}.
$$

Here, $(E_0)$ has units of energy, while $(\Theta_0)$ has units of velocity squared. The corresponding reference thermal speed is

$$
\boxed{c_{\mathrm{th},0}=\sqrt{\Theta_0} =\sqrt{\frac{k_B T_0}{m_0}}.}
$$

For a gas with molecular mass $(m)$ and local temperature $(T)$,

$$
\Theta=\frac{k_B T}{m} =\Theta_0\frac{T}{T_0}\frac{m_0}{m}.
$$

For a single species with $(m=m_0)$, this reduces to $(\Theta=\Theta_0 T/T_0)$.

The Maxwell–Boltzmann distribution, normalized to mass density, becomes

$$
f^{eq}(\boldsymbol{\xi}) = \frac{\rho}{(2\pi\Theta)^{D/2}} \exp\!\left[ -\frac{|\boldsymbol{\xi}-\mathbf u|^2}{2\Theta} \right].
$$

Consequently,

$$
p=\rho\Theta, \qquad e_{\mathrm{tr}}=\frac{D}{2}\Theta,
$$

where $(e_{\mathrm{tr}})$ is the specific translational internal energy.

Be precise about “sound speed”: for this ideal gas, the **isothermal** sound speed satisfies $(c_T^2=\Theta)$, whereas the **adiabatic** sound speed satisfies $(c_a^2=\gamma\Theta )$.

**2. Express this energy in lattice units**

Let the fixed streaming velocity scale be

$$
c_\ell=\frac{\Delta x}{\Delta t},
\qquad
\boldsymbol{\xi}_i=c_\ell\mathbf e_i,
$$

where $(\mathbf e_i)$ contains the lattice’s integer velocity components. Define the dimensionless variance

$$
\boxed{\theta=\frac{\Theta}{c_\ell^2} = \frac{k_B T_0}{m_0} \left(\frac{\Delta t}{\Delta x}\right)^2 \frac{T}{T_0}\frac{m_0}{m}. }
$$

Thus, at the reference state,

$$
\theta_0= \frac{k_B T_0}{m_0} \left(\frac{\Delta t}{\Delta x}\right)^2.
$$

This $(\theta)$ is the quantity you would identify with the squared isothermal sound speed **in lattice units**.

The distinction between physical and lattice units matters: changing $(T_0$) or $(m_0$) can be accommodated by changing $(\Delta t/\Delta x$), leaving $(\theta_0$) and the standard weights unchanged.

**3. For fixed velocities, derive energy-dependent weights from moment matching**

If you want $(\theta$) to vary while keeping $(\mathbf e_i$) fixed, use the Gaussian

$$
W_\theta(\mathbf v)=\frac{1}{(2\pi\theta)^{D/2}} \exp\!\left(-\frac{|\mathbf v|^2}{2\theta}\right)
$$

and require

$$
\boxed{\sum_i w_i(\theta)P(\mathbf e_i) = \int_{\mathbb R^D}W_\theta(\mathbf v)P(\mathbf v)\,d\mathbf v }
$$

for every polynomial $(P$) up to the required quadrature degree.

The first conditions are

$$
\sum_i w_i=1, \qquad \sum_i w_i e_{i\alpha}=0,
$$

$$
\sum_i w_i e_{i\alpha}e_{i\beta} =\theta\delta_{\alpha\beta},
$$

$$
\sum_i w_i e_{i\alpha}e_{i\beta}e_{i\gamma}=0,
$$

$$
\sum_i w_i e_{i\alpha}e_{i\beta}e_{i\gamma}e_{i\delta} = \theta^2 \left( \delta_{\alpha\beta}\delta_{\gamma\delta} +\delta_{\alpha\gamma}\delta_{\beta\delta} +\delta_{\alpha\delta}\delta_{\beta\gamma} \right).
$$

This is the direct extension of the lecture’s quadrature construction. Your desired dependence is then

$$
w_i=w_i\!\left( \frac{k_B T}{m} \frac{\Delta t^2}{\Delta x^2} \right).

**4. Why the standard D1Q3 or D2Q9 lattice is restrictive**

For D1Q3, take $(e_i=\{0,+1,-1\}$), with equal weights $(w_1$) for the moving velocities.

Normalization and the second moment give

$$
w_0+2w_1=1,
\qquad
2w_1=\theta,
$$

hence

$$
w_0=1-\theta,
\qquad
w_{\pm1}=\frac{\theta}{2}.
$$

These appear to allow variable $(\theta$). However, the fourth Gaussian moment also requires

$$
2w_1=3\theta^2.
$$

Combining the second and fourth moments gives

$$
\theta=3\theta^2
\quad\Longrightarrow\quad
\boxed{\theta=\frac13}
$$

for a nondegenerate Gaussian.

The same restriction holds for standard D2Q9 because every component satisfies $(e_{i\alpha}^4=e_{i\alpha}^2$).

Therefore, **you cannot freely vary the Gaussian variance on standard D2Q9 while retaining exact second- and fourth-order Gaussian moments merely by changing its weights**.

**5. A concrete extension: D1Q5 with variable weights**

Consider instead

$$
e_i=\{0,\pm1,\pm2\}.
$$

Let $(w_1$) be the weight of **each** $(\pm1$) velocity and $(w_2$) the weight of **each** $(\pm2$) velocity. Match moments through degree four:

$$
\begin{aligned}
w_0+2w_1+2w_2&=1,\\
2w_1+8w_2&=\theta,\\
2w_1+32w_2&=3\theta^2.
\end{aligned}
$$

Solving gives

$$
\boxed{
\begin{aligned}
w_0(\theta)&=1-\frac54\theta+\frac34\theta^2,\\
w_{\pm1}(\theta)&=\frac{\theta(4-3\theta)}{6},\\
w_{\pm2}(\theta)&=\frac{\theta(3\theta-1)}{24}.
\end{aligned}
}
$$

All weights are nonnegative for

$$
\boxed{\frac13\leq\theta\leq\frac43.}
$$

They are strictly positive inside this interval. At $(\theta=1/3$), the outer weights vanish and the construction reduces to standard D1Q3.

A Cartesian product gives corresponding constructions in higher dimensions:

$$
\begin{aligned}
D2Q25:\quad
w_{ij}(\theta)&=w_i(\theta)w_j(\theta),\\
D3Q125:\quad
w_{ijk}(\theta)&=w_i(\theta)w_j(\theta)w_k(\theta).
\end{aligned}
$$

These are **illustrative quadratures matching the stated moments**, not yet complete thermal LBM schemes. Higher moments needed for the intended energy and heat-flux equations must also be checked.

**6. Connect this to your Hermite formulation**

With variance $(\theta$), the first Hermite tensors in lattice coordinates become

$$
H^{(0)}=1,
\qquad
H^{(1)}_\alpha=e_\alpha,
\qquad
H^{(2)}_{\alpha\beta}
=e_\alpha e_\beta-\theta\delta_{\alpha\beta},
$$

$$
H^{(3)}_{\alpha\beta\gamma}
=
e_\alpha e_\beta e_\gamma
-\theta\left(
e_\alpha\delta_{\beta\gamma}
+e_\beta\delta_{\alpha\gamma}
+e_\gamma\delta_{\alpha\beta}
\right).
$$

Using this convention, the discrete expansion has the form

$$
\boxed{
f_i
=
w_i(\theta)
\sum_{n=0}^{N}
\frac{
a^{(n)}:H^{(n)}(\mathbf e_i;\theta)
}{
n!\,\theta^n
}.
}
$$

Here, $(N$) is the chosen Hermite truncation order; it is not generally $(q-1$) in multiple dimensions. The lattice must support the associated discrete orthogonality.

If $(\theta=\theta(\mathbf x,t)$), both the weights and this basis vary locally. A full derivation must account for that variation and include an energy evolution equation; replacing $(c_s^2$) and $(w_i$) in an existing isothermal implementation alone is insufficient.

For your proposed scheme, the central construction is therefore

$$
\boxed{
(T_0,m_0,T,m)
\;\longrightarrow\;
\theta
\;\longrightarrow\;
\text{moment constraints}
\;\longrightarrow\;
w_i(\theta).
}
$$
