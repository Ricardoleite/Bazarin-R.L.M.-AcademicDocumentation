# Generalized Difference Finite Scheme for Balance Equations derived from Lattice Boltzmann

In the remainder of this section, we develop a finite difference scheme interpretation of balance equation in LBM framework:

Given the LBM formulation for the diffusive equation:

$$
f_i( x_{\alpha} + e_{i,\alpha} \delta t, t+\delta t) = f_i(x_{\alpha}, t)  -\left( \frac{ f_{i} - f_{i}^{eq} }{ \tau } \right) \qquad \rightarrow \qquad f_i( x_{\alpha} + e_{i,\alpha} \delta t, t+\delta t) = f_i(x_{\alpha}, t)  - \omega\left( f_{i} - f_{i}^{eq} \right), 
$$(LB-fdm-df-Eq)

where $f^{eq}_{i}$ is the equilibrium distribution function that define the moments conserved also called by equilibrium moments in a expansion of Hermite bases:

$$
f^{eq}_{i}(m^{(0)},...,m^{(n)})=w_{i}\displaystyle\sum_{\theta=0}^{\infty} \displaystyle\frac{ a^{(\theta)}(\phi,u_{\alpha},\Theta)\mathcal{H}^{(\theta)} }{\theta! c_{s}^{2\theta} },
$$(general-df-fdm-feq-Eq)

where $a^{(\theta)}$ is a tensor of order $\theta$ called expansion coefficient, $\mathcal{H}^{(\theta)}$ is Hermite polynomial tensor of order $\theta$. The terms $m^{(0)}(x_{\alpha}, t)$ up to $m^{(n)}(x_{\alpha}, t)$ are examples of a consecutive list of $n$ conservative moments that can be recovered by normal product of distribution funcion with specific polynomes $\mathcal{C}^{(\theta)}$ due to its ortogonality with Hermite polynomes. So, as example

$$
\sum_{i}f_{i}\mathcal{C}^{(0)}=\displaystyle\sum_{i}f_{i}^{eq}\mathcal{C}^{(0)}=m^{(0)}, \qquad \sum_{i}f_{i}\mathcal{C}^{(1)}=\displaystyle\sum_{i}f_{i}^{eq}\mathcal{C}^{(1)}=m^{(1)}, \qquad ... \qquad, \qquad \textrm{ and } \qquad \sum_{i}f_{i}\mathcal{C}^{(\theta)}=\displaystyle\sum_{i}f_{i}^{eq}\mathcal{C}^{(\theta)}=m^{(\theta)}. 
$$(general-conserved-moments-fdm-fi-df)

The expansion coefficients for each order are recovered by the product of distribution funcion with the respective Hermite polynomial tensor of order, as given in the equation bellow up to the third-order:

$$
\displaystyle\sum_{i}f_{i}\mathcal{H}^{(0)}=\displaystyle\sum_{i}f_{i}^{eq}\mathcal{H}^{(0)}=a^{(0)}, \qquad \displaystyle\sum_{i}f_{i}^{eq}\mathcal{H}^{(1)}_{\alpha} =a^{(1)}_{\alpha}, \qquad \displaystyle \sum_{i} f_{i}^{eq} \mathcal{H}^{(2)}_{\alpha\beta} =a^{(2)}_{\alpha\beta} ,\qquad \textrm{ and } \quad \quad \displaystyle\sum_{i} f_{i}^{eq} \mathcal{H}^{(3)}_{\alpha\beta\gamma} = a^{(3)}_{\alpha\beta\gamma}. 
$$(general-moments-fdm-fi-df)

For simplification we redefine the notation for $f_i( x_{\alpha} + e_{i,\alpha} \delta t, t+\delta t)=f_{i,x+c_{i}}^{t+1}$. Reinterpreting the Eq. {eq}`LB-fdm-df-Eq` in pre-streaming moment, we apply a shift in $c_{i}$:

$$
f_{i,x}^{t+1}  = f_{i,x-c_{i}}^{t}  - \omega \left( f_{i,x-c_{i}}^{t} - f_{i,x-c_{i}}^{eq,t} \right).
$$(pre-fdm-LB-df-Eq)

Multiplying by the $n$-order polynome of Hermite, i.e., $\mathcal{H}^{(n)}$, that correct recover a conserved moment and summing the Eq. {eq}`pre-fdm-LB-df-Eq` over all lattice direction, we have:

$$
\displaystyle\sum_{i}\Big( f_{i,x}^{t+1}\mathcal{C}^{(n)} \Big) = \displaystyle\sum_{i}\Big( f_{i,x-c_{i}}^{t}\mathcal{C}^{(n)}  - \omega \left( f_{i,x-c_{i}}^{t}\mathcal{C}^{(n)} - f_{i,x-c_{i}}^{eq,t}\mathcal{C}^{(n)} \right) \Big) \qquad \rightarrow  \qquad m_{x}^{(n)t+1} = \displaystyle\sum_{i}\Big( (1-\omega) f_{i,x-c_{i}}^{t}\mathcal{C}^{(n)}  + \omega f_{i,x-c_{i}}^{eq,t}\mathcal{C}^{(n)} \Big),
$$

given that we know all the moment described in the equilibrium distribution function, we have

$$
m_{x}^{(n)t+1} = \textcolor{green}{\sum_{i}\left(\omega f_{i,x-c_{i}}^{eq,t}\mathcal{C}^{(n)} \right)} + \sum_{i}\left( (1-\omega) f_{i,x-c_{i}}^{t}\mathcal{C}^{(n)}  \right),
$$

using the Eq. {eq}`general-conserved-moments-fdm-fi-df` to replace $f_{i,x-c_{i}}^{t}$ in the above equation:

$$
m_{x}^{(n)t+1} = \textcolor{green}{\sum_{i}\left(\omega f_{i,x-c_{i}}^{eq,t}\mathcal{C}^{(n)} \right)} + \sum_{i}\left( (1-\omega) f_{i,x-c_{i}}^{t}\mathcal{C}^{(n)}  \right) \qquad \rightarrow  \qquad m_{x}^{(n)t+1} = \textcolor{green}{\sum_{i}\left(\omega f_{i,x-c_{i}}^{eq,t}\mathcal{C}^{(n)} + (1-\omega)m_{x-c_{i}}^{(n)t} \right)} - (1-\omega) \sum_{i}\left( \sum_{k\neq i} \left(f_{k,x-c_{i}}^{t}\mathcal{C}^{(n)}  \right) \right).
$$

Repeting the process of shift in $c_{i}$ the unknow distribution function and write it terms of pre collisional function, we have:

$$
m_{x}^{(n)t+1} = \textcolor{green}{\sum_{i}\left(\omega f_{i,x-c_{i}}^{eq,t}\mathcal{C}^{(n)} + (1-\omega)m_{x-c_{i}}^{(n)t} \right)} - (1-\omega) \sum_{i}\left( \sum_{k\neq i} \left(\left( (1-\omega)f_{k,x-c_{i}-c_{k}}^{t-1}  + \omega f_{k,x-c_{i}-c_{k}}^{eq,t-1} \right)\mathcal{C}^{(n)}  \right) \right),
$$

$$
m_{x}^{(n)t+1} = \textcolor{green}{\sum_{i}\left(\omega f_{i,x-c_{i}}^{eq,t}\mathcal{C}^{(n)} + (1-\omega)m_{x-c_{i}}^{(n)t} \right)} - \textcolor{green}{(1-\omega) \sum_{i}\left( \sum_{k\neq i} \left( \omega f_{k,x-c_{i}-c_{k}}^{eq,t-1} \mathcal{C}^{(n)}  \right) \right)} - (1-\omega) \sum_{i}\left( \sum_{k\neq i} \left( (1-\omega)f_{k,x-c_{i}-c_{k}}^{t-1} \mathcal{C}^{(n)}  \right) \right),
$$

$$
m_{x}^{(n)t+1} = \textcolor{green}{\sum_{i}\left(\omega f_{i,x-c_{i}}^{eq,t}\mathcal{C}^{(n)} + (1-\omega)m_{x-c_{i}}^{(n)t} \right)} + \textcolor{green}{(1-\omega) \sum_{i}\left( \sum_{k\neq i} \left( \omega f_{k,x-c_{i}-c_{k}}^{eq,t-1} + (1-\omega)m_{x-c_{i}-c_{k}}^{(n)t-1} \mathcal{C}^{(n)}  \right) \right)} + (1-\omega)^{2} \sum_{i}\left( \sum_{k\neq i} \left( \sum_{j\neq k} \left( f_{j,x-c_{i}-c_{k}}^{t-1} \mathcal{C}^{(n)}  \right) \right)\right).
$$

Considering that possible reapeat $n$ times the process shift in $c_{i}$ the unknow distribution function and write it terms of pre collisional function, and replacing the lattice index $i,k,j$ by $i_{0},i_{1},i_{2}$, respectively, to create a order $i_{n}$ index, we will have:

$$
m_{x}^{(n)t+1} = \textcolor{green}{\sum_{i_{0}}\left(\omega f_{i_{0},x-c_{i_{0}}}^{eq,t}\mathcal{C}^{(n)} + (1-\omega)m_{x-c_{i_{0}}}^{(n)t} \right)} + \textcolor{green}{(1-\omega) \sum_{i_{0}}\left( \sum_{i_{1}\neq i_{0}} \left( \omega f_{i_{1},x-c_{i_{0}}-c_{i_{1}}}^{eq,t-1} + (1-\omega)m_{x-c_{i_{0}}-c_{i_{1}}}^{(n)t-1} \mathcal{C}^{(n)}  \right) \right)} + ... +(1-\omega)^{n-1} \sum_{i_{0}}\left( \sum_{i_{1}\neq i_{0}} ... \left( \sum_{i_{n-1}\neq i_{n-2}} \left( f_{i_{n-1},x-c_{i_{0}}-c_{i_{1}}-...-c_{i_{n-2}}}^{t-n-2} \mathcal{C}^{(n)}  \right) \right)\right).
$$

Given that you are employing a lattice D$d$Q$n$, the next step of shift in $c_{i_{n-1}}$ the unknow distribution function and write it terms of pre collisional function after $n$ repeated steps will result in cancel due to distribution function be shifted in all possible directions, i.e.,:

$$
(1-\omega)^{n-1} \sum_{i_{0}}\left( \sum_{i_{1}\neq i_{0}} ... \left( \sum_{i_{n-1}\neq i_{n-2}} \left( f_{i_{n-1},x-c_{i_{0}}-c_{i_{1}}-...-c_{i_{n-2}}}^{t-n-2} \mathcal{C}^{(n)}  \right) \right)\right) = (1-\omega)^{n-1} \sum_{i_{0}}\left( \sum_{i_{1}\neq i_{0}} ... \left( \sum_{i_{n-1}\neq i_{n-2}} \left( \left((1-\omega)f_{i_{n-1},x-c_{i_{0}}-c_{i_{1}}-...-c_{i_{n-2}}-c_{i_{n-1}}}^{t-n-1} + \omega f_{i_{n-1},x-c_{i_{0}}-c_{i_{1}}-...-c_{i_{n-2}}-c_{i_{n-1}}}^{eq,t-n-1} \right)\mathcal{C}^{(\theta)}  \right) \right)\right)
$$

$$
(1-\omega)^{n-1} \sum_{i_{0}}\left( \sum_{i_{1}\neq i_{0}} ... \left( \sum_{i_{n-1}\neq i_{n-2}} \left( \left((1-\omega)f_{i_{n-1},x-c_{i_{0}}-c_{i_{1}}-...-c_{i_{n-2}}-c_{i_{n-1}}}^{t-n-1} + \omega f_{i_{n-1},x-c_{i_{0}}-c_{i_{1}}-...-c_{i_{n-2}}-c_{i_{n-1}}}^{eq,t-n-1} \right)\mathcal{C}^{(\theta)}  \right) \right)\right)= \textcolor{green}{(1-\omega)^{n-1} m_{x}^{(\theta)t-n-1}}
$$
