# Generalized Difference Finite Scheme for Balance Equations derived from Lattice Boltzmann

In the remainder of this section, we develop a generalized finite difference scheme (FDS) interpretation of balance equation within the LBM framework. Initially, we dedvelop the analysis under a single-relaxation time (SRT) framework and subsequently we extendo multi-relaxation times (MRT) framework. In both case are considered a generalized lattice $DdQq$.

Given the lattice Boltzmann (LB) equatioin in a SRT formulation:

$$
f_i( x_{\alpha} + c_{i} \delta t, t+\delta t) = f_i(x_{\alpha}, t)  -\left( \frac{ f_{i} - f_{i}^{eq} }{ \tau } \right) \qquad \rightarrow \qquad f_i^{'} = f_i  - \omega\left( f_{i} - f_{i}^{eq} \right), 
$$(LB-fdm-df-Eq)

where $f^{eq}_{i}$ is the equilibrium distribution function that define the moments conserved also called by equilibrium moments in a expansion of Hermite bases:

$$
f^{eq}_{i}\left(m^{\mathcal{C}(0)},...,m^{\mathcal{C}(n)}\right)=w_{i}\displaystyle\sum_{\theta=0}^{N} \displaystyle\frac{ m^{eq,(\theta)}\left(m^{\mathcal{C}(0)},...,m^{\mathcal{C}(n)}\right)\mathcal{H}^{(\theta)} }{\theta! c_{s}^{2\theta} },
$$(general-df-fdm-feq-Eq)

where $m^{eq,(\theta)}$ is a tensor of order $\theta$ called equilibrium expansion coefficient or Hermite equilibrium moments, $\mathcal{H}^{(\theta)}$ is Hermite polynomial tensor of order $\theta$, and $N$ is the lattice order. The terms $m^{\mathcal{C}(0)}(x_{\alpha}, t)$ up to $m^{\mathcal{C}(n)}(x_{\alpha}, t)$ are examples of a consecutive list of $n$ conservative moments that can be recovered by normal product of distribution funcion with specific polynomes $\mathcal{C}_{c_{i}}^{(\theta)}$ due to its ortogonality with Hermite polynomes. So, as example

$$
\begin{array}{c}
\sum_{i}f_{i}\mathcal{C}_{c_{i}}^{(0)}=\displaystyle\sum_{i}f_{i}^{eq}\mathcal{C}_{c_{i}}^{(0)}=m^{\mathcal{C}(0)}, \qquad \sum_{i}f_{i}\mathcal{C}_{c_{i}}^{(1)}=\displaystyle\sum_{i}f_{i}^{eq}\mathcal{C}_{c_{i}}^{(1)}=m^{\mathcal{C}(1)},\\
\sum_{i}f_{i}\mathcal{C}_{c_{i}}^{(2)}=\displaystyle\sum_{i}f_{i}^{eq}\mathcal{C}_{c_{i}}^{(2)}=m^{\mathcal{C}(2)}, \qquad ... \qquad, \textrm{ and } \qquad \sum_{i}f_{i}\mathcal{C}_{c_{i}}^{(\theta)}=\displaystyle\sum_{i}f_{i}^{eq}\mathcal{C}_{c_{i}}^{(\theta)}=m^{\mathcal{C}(\theta)}. 
\end{array}
$$(general-conserved-moments-fdm-fi-df)

The expansion coefficients for each order are recovered by the product of distribution funcion with the respective Hermite polynomial tensor of order, as given in the equation bellow up to the third-order:

$$
\begin{array}{c}
\displaystyle\sum_{i}f_{i}\mathcal{H}^{(0)}=\displaystyle\sum_{i}f_{i}^{eq}\mathcal{H}^{(0)}=m^{eq,(0)}, \qquad \displaystyle\sum_{i}f_{i}^{eq}\mathcal{H}^{(1)}_{\alpha} =m^{eq,(1)}_{\alpha}, \qquad \displaystyle \sum_{i} f_{i}^{eq} \mathcal{H}^{(2)}_{\alpha\beta} =m^{eq,(2)}_{\alpha\beta},\\
\displaystyle\sum_{i} f_{i}^{eq} \mathcal{H}^{(3)}_{\alpha\beta\gamma} = m^{eq,(3)}_{\alpha\beta\gamma}, \qquad ... \qquad, \textrm{ and } \qquad \displaystyle\sum_{i} f_{i}^{eq} \mathcal{H}^{(q)}_{\alpha\beta\gamma} = m^{eq,(q)}_{\alpha\beta\gamma}.
\end{array}
$$(general-moments-fdm-fi-df)

## FDS of Multi-Relaxation Time (MRT) LBM framework

Given the lattice Boltzmann (LB) equatioin in a generalized MRT formulation:

$$
f_i( x_{\alpha} + e_{i,\alpha} \delta t, t+\delta t) = f_i(x_{\alpha}, t)  - (\textbf{M}^{-1}\textbf{S}\textbf{M})_{ik}\left( f_{k} - f_{k}^{eq}  \right), 
$$(MRT-LB-conv-df-Eq)

where $\textbf{M}$ is matrix of Hermite moments, $\textbf{S}$ is the diagonal matrix of relaxation times, and $\textbf{M}^{-1}$ is the inverse matrix of $\textbf{M}$. Considering the problem description in a lattice $D1Qq$, we have:

$$
\textbf{M} =\begin{pmatrix} \mathcal{H}^{(0)} \\ \mathcal{H}^{(1)}_{\alpha_{1}} \\ \mathcal{H}^{(2)}_{\alpha_{1}\alpha_{2}} \\ \vdots \\ \mathcal{H}^{(q-1)}_{\alpha_{1}\alpha_{2}\cdots\alpha_{q-1}}\end{pmatrix} ,  \qquad \qquad  \textbf{S} = \begin{pmatrix} s_0 & 0   & 0 & \cdots & 0 \\ 0 & s_1 & 0 & \cdots & 0 \\ 0   & 0   & s_2 & \cdots & 0 \\ \vdots & \vdots & \vdots & \ddots & \vdots \\ 0 & 0 & 0 & \cdots & s_{q-1} \end{pmatrix}.
$$

The equilibrium distribution function is given by Eq. {eq}`general-df-fdm-feq-Eq`, and the distribution function is defined by

$$
f_{i} = w_{i}\displaystyle\sum_{\theta=0}^{N}\frac{m^{(\theta)}\mathcal{H}^{(\theta)}_{c_{i}}}{\theta!c_{s}^{2\theta}}.
$$

where $m^{(\theta)}$ is the Hermite moments also called by expansion coefficients.

Rewritting the collision term:

$$
(\textbf{M}^{-1}\textbf{S}\textbf{M})_{ik}\left( f_{k} - f_{k}^{eq}  \right) = f_{i}^{S} - f_{i}^{S,eq}, 
$$

by transform the distribution function to moments space relaxing each moment and inverting transformation, we have

$$
f_{i}^{S} = w_{i}\displaystyle\sum_{\theta=0}^{q-1}\frac{s_{\theta}m^{(\theta)}\mathcal{H}^{(\theta)}}{\theta!c_{s}^{2\theta}}=w_{i}\left( s_{0} m^{(0)}\mathcal{H}^{(0)} + \frac{s_{1}m^{(1)}_{\alpha_{1}}\mathcal{H}^{(1)}_{\alpha_{1}}}{c_{s}^{2}} + \cdots + \frac{s_{q-1}m^{(q-1)}_{\alpha_{1}\cdots \alpha_{q-1}}\mathcal{H}^{(q-1)}_{\alpha_{1}\cdots \alpha_{q-1}}}{(q-1)!c_{s}^{2(q-1)}} \right).
$$

considering a one-dimensional lattice. It can be extended for d-dimensional lattices described for full projection of one-dimensional lattices.

For simplification we redefine the notation for $f_i( x_{\alpha} + e_{i,\alpha} \delta t, t+\delta t)=f_{i,x+c_{i}}^{t+1}$ and remove the index tensor order notation of Hermite polynomes and moments. Reinterpreting the Eq. {eq}`LB-fdm-df-Eq` in pre-streaming moment, we apply a shift in $c_{i}$:

$$
\begin{array}{rl}
f_{i,x_{\alpha}}^{t+1} = & f_{i,x_{\alpha}-c_{i}}^{t} - \left( f_{i,x_{\alpha}-c_{i}}^{S,t} - f_{i,x_{\alpha}-c_{i}}^{S,eq,t} \right) =  f_{i,x_{\alpha}-c_{i}}^{S,eq,t} + f_{i,x_{\alpha}-c_{i}}^{I-S,t} ,
\end{array}
$$

where $f_{i,x_{\alpha}-c_{i}}^{I-S,t}$ is given by:

$$
f_{i}^{I-S} = w_{i}\displaystyle\sum_{\theta=0}^{q-1}\frac{(1-s_{\theta})m^{(\theta)}\mathcal{H}^{(\theta)}}{\theta!c_{s}^{2\theta}} = w_{i}\left( (1-s_{0}) m^{(0)}\mathcal{H}^{(0)} + \frac{ (1-s_{1})m^{(1)}_{\alpha_{1}}\mathcal{H}^{(1)}_{\alpha_{1}}}{c_{s}^{2}} + \cdots + \frac{(1-s_{q-1})m^{(q-1)}_{\alpha_{1}\cdots \alpha_{q-1}}\mathcal{H}^{(q-1)}_{\alpha_{1}\cdots \alpha_{q-1}}}{(q-1)!c_{s}^{2(q-1)}} \right).
$$

However, we will also use a similar notation to the one presented in {cite:t}`bellotti2022finite`, but we define matricial products that keep terms in the space of moments and space of distribution functions, represented by super-script $m$ and $f$, respectively. Therefore, we have $T_{\Delta x}^{c_{i}}I= diag(T_{\Delta x}^{c_{0}},\cdots, T_{\Delta x}^{c_{q-1}})$ as the shift matrix and the terms:

$$
T_{i}^{m}=M(T_{\Delta x}^{c_{i}}I)M^{-1}, \qquad Q_{i}^{m}=T_{i}^{m}S, \qquad P_{i}^{m}=T_{i}^{m}(I-S),
$$

T_{i}^{f}=M^{-1}(T_{\Delta x}^{c_{i}}I)M, \qquad Q_{i}^{f}=T_{i}^{f}S, \qquad \textrm{and}\qquad P_{i}^{f}=T_{i}^{f}(I-S).
$$ -->

Q_{i}^{f}=(T_{\Delta x}^{c_{i}}I) M^{-1}SM, \qquad \textrm{and}\qquad P_{i}^{f}=(T_{\Delta x}^{c_{i}}I) M^{-1}(I-S)M.

The main task of the proposed notation is reduce the size of matricial operations developed in sequence of presence analysis. Using the defined notation, we rewrite

f_{i,x_{\alpha}}^{t+1} = & f_{i,x_{\alpha}-c_{i}}^{t} - \left( f_{i,x_{\alpha}-c_{i}}^{S,t} - f_{i,x_{\alpha}-c_{i}}^{S,eq,t} \right) = Q_{i}^{f} f_{i,x_{\alpha}}^{eq,t} + P_{i}^{f} f_{i,x_{\alpha}}^{t}.

## Continuation: exact elimination of nonconserved moments for D1Qq

This continuation starts from the final MRT expansion above. The original cells are retained as supplied; the corrections below apply when using their formulas. The derivation preserves $f_i^{S,eq}$, $f_i^{I-S}$, $P^f$, $Q^f$, $P^m$, $Q^m$, $\mathcal H^{(\theta)}$, $m^{\mathcal C(n)}$, and the original meaning of $\gamma_0$. New symbols are introduced only to organize the elimination.



### 1. Assumptions and notation corrections

Work in lattice units with $i=0,\ldots,q-1$ and distinct one-dimensional velocities $c_i$. Define the backward spatial shift by

$$
T_{\Delta x}^{c_i}g(x_\alpha)=g(x_\alpha-c_i),\qquad
T_{\Delta x}^{c_i}T_{\Delta x}^{c_k}=T_{\Delta x}^{c_i+c_k}.
$$

In physical coordinates, the displacement in this definition is $c_i\Delta x$ when $c_i$ is an integer lattice velocity. Throughout the formulas below, $x_\alpha-c_i$ uses lattice coordinates.

Assume $M$, $S$, $w_i$, and $c_s$ are constant in space and time. Equilibrium may be nonlinear, but it must be a specified local function of the conserved moments. The derivation applies on the infinite lattice or a periodic domain, and to interior stencils on bounded domains. Boundary closure requires separate treatment.

* The entries $s_\theta$ of $S$ are relaxation **rates**.
* In the section with $\mathcal C_{c_i}^{(0)}=1$, the left-hand moment is $m^{\mathcal C(0)}$, not a general $m^{\mathcal C(n)}$.
* All Hermite denominators in the preceding expansion must read $\theta!c_s^{2\theta}$, including the residual sums.
* Nested Hermite sums require independent indices, for example $\theta$ and $\eta$.
* A matrix such as $Q^f$ acts on the full population vector: the component equation is $f_i^{t+1}=\sum_k(Q^f)_{ik}f_k^{eq,t}+\sum_k(P^f)_{ik}f_k^t$.
* The coefficients $\gamma_r$ below are spatial **operators**. Write $\gamma_r m$, with the operator acting on the field, rather than treating it as a numerical multiplier.

For a generic D1Qq lattice, set

$$
M_{\theta i}=\mathcal H_{c_i}^{(\theta)},\qquad
m^{(\theta)}=\sum_i M_{\theta i}f_i,\qquad
f_i=\sum_{\theta=0}^{q-1}(M^{-1})_{i\theta}m^{(\theta)}.
$$

The particular inverse used in the original notebook,

$$
(M^{-1})_{i\theta}=\frac{w_i\mathcal H_{c_i}^{(\theta)}}{\theta!c_s^{2\theta}},
$$

is exact only if the chosen lattice satisfies the full discrete orthogonality relation

$$
\sum_i w_i\mathcal H_{c_i}^{(\theta)}\mathcal H_{c_i}^{(\eta)}
=\theta!c_s^{2\theta}\delta_{\theta\eta},\qquad 0\le\theta,\eta\le q-1.
$$

This must not be assumed for every integer-velocity D1Qq. The development below uses the actual $M^{-1}$ and therefore remains valid when that Hermite reconstruction formula is unavailable. Reconstructing arbitrary nonequilibrium populations requires all $q$ independent moments; a truncated equilibrium expansion is allowed if it defines the intended equilibrium.



### 2. Compress the collision coefficients, without changing their meaning

Introduce the constant matrices

$$
A_{ik}=\big[M^{-1}(I-S)M\big]_{ik},\qquad
D_{ik}=\big[M^{-1}SM\big]_{ik}=\delta_{ik}-A_{ik}.
$$

Under the discrete orthogonality condition above, your coefficient becomes

$$
\boxed{A_{ik}=w_i\sum_{\theta=0}^{q-1}
\frac{(1-s_\theta)\mathcal H_{c_i}^{(\theta)}\mathcal H_{c_k}^{(\theta)}}{\theta!c_s^{2\theta}}.}
$$

Thus $f_i^{I-S}=\sum_k A_{ik}f_k$ and $f_i^{S,eq}=\sum_k D_{ik}f_k^{eq}$. With

$$
\mathsf T=\operatorname{diag}(T_{\Delta x}^{c_0},\ldots,T_{\Delta x}^{c_{q-1}}),
$$

your operators are

$$
P^f=\mathsf T A,\quad Q^f=\mathsf T D,\quad
P^m=MP^fM^{-1},\quad Q^m=MQ^fM^{-1}.
$$

The subscript $i$ on the full matrices in the original notation is suppressed: their entries already carry their velocity indices. In components,

$$
(P^f)_{ik}=A_{ik}T_{\Delta x}^{c_i},\qquad
f_{i,x_\alpha}^{t+1}=f_{i,x_\alpha-c_i}^{S,eq,t}
+\sum_k A_{ik}f_{k,x_\alpha-c_i}^t.
$$

For compactness only, write $P=P^f$, $Q=Q^f$, $\boldsymbol f=(f_0,\ldots,f_{q-1})^T$, and $\boldsymbol 1^T=(1,\ldots,1)$. Then

$$
\boldsymbol f^{t+1}=P\boldsymbol f^t+Q\boldsymbol f^{eq,t},\qquad
m^{\mathcal C(0)t}=\boldsymbol 1^T\boldsymbol f^t.
$$



### 3. First elimination: recover the notebook's gamma_0

Sum the component equation over $i$, split $k=i$ from $k\ne i$, and insert

$$
f_{i,x_\alpha-c_i}^t=m_{x_\alpha-c_i}^{\mathcal C(0)t}
-\sum_{k\ne i}f_{k,x_\alpha-c_i}^t.
$$

This gives exactly

$$
\boxed{
\begin{aligned}
m_{x_\alpha}^{\mathcal C(0)t+1}
={}&\sum_i f_{i,x_\alpha-c_i}^{S,eq,t}
+\gamma_0m_{x_\alpha}^{\mathcal C(0)t}
+\sum_i\sum_{k\ne i}(A_{ik}-A_{ii})f_{k,x_\alpha-c_i}^t,\\
\gamma_0={}&\sum_i A_{ii}T_{\Delta x}^{c_i}=\operatorname{tr}P.
\end{aligned}}
$$

The Hermite form of $A_{ik}-A_{ii}$ is precisely the coefficient with $\mathcal H_{c_k}^{(\theta)}-\mathcal H_{c_i}^{(\theta)}$ in your last cell, after correcting the denominators.

Define $B_0=I$ and $B_1=P-\gamma_0I$. The same equation is

$$
m^{\mathcal C(0)t+1}=\gamma_0m^{\mathcal C(0)t}
+\boldsymbol1^TQ\boldsymbol f^{eq,t}
+\boldsymbol1^TB_1\boldsymbol f^t.
$$

The last term is the remaining unknown population contribution.



### 4. Second elimination in explicit shifted-index notation

In the remaining sum, use

$$
f_{k,x_\alpha-c_i}^t=f_{k,x_\alpha-c_i-c_k}^{S,eq,t-1}
+\sum_l A_{kl}f_{l,x_\alpha-c_i-c_k}^{t-1}.
$$

The equilibrium part becomes

$$
\mathcal E_1^{t-1}=\sum_i\sum_{k\ne i}(A_{ik}-A_{ii})
f_{k,x_\alpha-c_i-c_k}^{S,eq,t-1}
=\boldsymbol1^TB_1Q\boldsymbol f^{eq,t-1}.
$$

Group the two ordered pairs $(i,k)$ and $(k,i)$ into $i<k$. With $c_{i,k}=c_i+c_k$, define

$$
\begin{aligned}
b_{ik,l}&=(A_{ik}-A_{ii})A_{kl}+(A_{ki}-A_{kk})A_{il},\\
d_{ik}&=A_{ii}A_{kk}-A_{ik}A_{ki}.
\end{aligned}
$$

The unknown part is $\sum_{i<k}\sum_l b_{ik,l}f_{l,x_\alpha-c_{i,k}}^{t-1}$. Direct substitution shows

$$
b_{ik,i}=b_{ik,k}=-d_{ik}.
$$

Therefore, using $f_i+f_k=m^{\mathcal C(0)}-\sum_{l\notin\{i,k\}}f_l$ at the same shifted point,

$$
\sum_l b_{ik,l}f_l=-d_{ik}m^{\mathcal C(0)}
+\sum_{l\notin\{i,k\}}(b_{ik,l}+d_{ik})f_l.
$$

The complete second-level equation is consequently

$$
\boxed{
\begin{aligned}
m_{x_\alpha}^{\mathcal C(0)t+1}
={}&\gamma_0m_{x_\alpha}^{\mathcal C(0)t}
+\gamma_1m_{x_\alpha}^{\mathcal C(0)t-1}
+\sum_i f_{i,x_\alpha-c_i}^{S,eq,t}
+\mathcal E_1^{t-1}\\
&+\sum_{i<k}\sum_{l\notin\{i,k\}}
(b_{ik,l}+d_{ik})f_{l,x_\alpha-c_{i,k}}^{t-1},\\
\gamma_1={}&-\sum_{i<k}d_{ik}T_{\Delta x}^{c_i+c_k}.
\end{aligned}}
$$

This is the direct continuation of your final cell. The unknown sum now excludes both shift indices. In MRT, replacing a remaining $f_l$ introduces every collision index again; the next exclusion is obtained by collecting and cancelling terms, not by assuming that the new collision index is automatically distinct.



### 5. Systematically organize every subsequent cancellation

The same second-level result follows by substituting $\boldsymbol f^t=P\boldsymbol f^{t-1}+Q\boldsymbol f^{eq,t-1}$ into the first-level residual:

$$
\boldsymbol1^TB_1\boldsymbol f^t
=\boldsymbol1^TB_1Q\boldsymbol f^{eq,t-1}
+\boldsymbol1^TB_1P\boldsymbol f^{t-1}.
$$

Set

$$
\gamma_1=\frac12\operatorname{tr}(PB_1)
=\frac12\left[\operatorname{tr}(P^2)-(\operatorname{tr}P)^2\right],
\qquad B_2=PB_1-\gamma_1I.
$$

Then $B_1P=B_2+\gamma_1I$, and the remaining population term is $\boldsymbol1^TB_2\boldsymbol f^{t-1}$. Expanding the traces gives exactly the pair-minor formula for $\gamma_1$ above.

For every $r=1,\ldots,q$, continue with

$$
\boxed{
B_0=I,\qquad
\gamma_{r-1}=\frac1r\operatorname{tr}(PB_{r-1}),\qquad
B_r=PB_{r-1}-\gamma_{r-1}I.
}
$$

All $B_r$ are polynomials in $P$, so $B_rP=PB_r$. This is the Faddeev-LeVerrier construction with signs and indices chosen to preserve your $\gamma_0$.

For example,

$$
\begin{aligned}
B_1&=P-\gamma_0I,\\
B_2&=P^2-\gamma_0P-\gamma_1I,\\
\gamma_2&=\frac13\operatorname{tr}(PB_2)
=\frac16\left[(\operatorname{tr}P)^3
-3\operatorname{tr}P\operatorname{tr}(P^2)+2\operatorname{tr}(P^3)\right],\\
B_3&=P^3-\gamma_0P^2-\gamma_1P-\gamma_2I.
\end{aligned}
$$

At the third level,

$$
\begin{aligned}
m^{\mathcal C(0)t+1}={}&\gamma_0m^{\mathcal C(0)t}
+\gamma_1m^{\mathcal C(0)t-1}+\gamma_2m^{\mathcal C(0)t-2}\\
&+\boldsymbol1^TQ\boldsymbol f^{eq,t}
+\boldsymbol1^TB_1Q\boldsymbol f^{eq,t-1}
+\boldsymbol1^TB_2Q\boldsymbol f^{eq,t-2}
+\boldsymbol1^TB_3\boldsymbol f^{t-2}.
\end{aligned}
$$



### 6. Induction and exact termination

After $r$ elimination levels, the identity is

$$
\boxed{
\begin{aligned}
m^{\mathcal C(0)t+1}
={}&\sum_{a=1}^{r}\gamma_{a-1}m^{\mathcal C(0)t+1-a}
+\sum_{a=0}^{r-1}\boldsymbol1^TB_aQ\boldsymbol f^{eq,t-a}
+\boldsymbol1^TB_r\boldsymbol f^{t+1-r}.
\end{aligned}}
$$

To pass from $r$ to $r+1$, substitute the LB update into the final term:

$$
\begin{aligned}
\boldsymbol1^TB_r\boldsymbol f^{t+1-r}
&=\boldsymbol1^TB_rQ\boldsymbol f^{eq,t-r}
+\boldsymbol1^TB_rP\boldsymbol f^{t-r}\\
&=\boldsymbol1^TB_rQ\boldsymbol f^{eq,t-r}
+\gamma_rm^{\mathcal C(0)t-r}
+\boldsymbol1^TB_{r+1}\boldsymbol f^{t-r}.
\end{aligned}
$$

For constant coefficients the entries of $P$ belong to the commutative algebra of spatial shifts. Its characteristic polynomial is

$$
\chi_P(\lambda)=\det(\lambda I-P)
=\lambda^q-\gamma_0\lambda^{q-1}-\gamma_1\lambda^{q-2}
-\cdots-\gamma_{q-1}.
$$

Cayley-Hamilton gives

$$
\boxed{B_q=P^q-\gamma_0P^{q-1}-\cdots-\gamma_{q-1}I=0.}
$$

Thus at $r=q$ the remaining nonequilibrium populations disappear identically. This requires $q-1$ further substitutions after the first-level equation, not an infinite expansion or an approximation.



### 7. Final scheme in conserved moments only

For one conserved moment, define

$$
\mathcal E_r[m^{\mathcal C(0)}]
=\boldsymbol1^TB_rQ\boldsymbol f^{eq}(m^{\mathcal C(0)}),\qquad 0\le r\le q-1.
$$

The exact finite-difference scheme is

$$
\boxed{
m_{x_\alpha}^{\mathcal C(0)t+1}
=\sum_{r=0}^{q-1}\left\{
\gamma_r m_{x_\alpha}^{\mathcal C(0)t-r}
+\mathcal E_r[m^{\mathcal C(0)t-r}]_{x_\alpha}
\right\}.
}
$$

Equivalently, without introducing $\mathcal E_r$,

$$
\boxed{
m^{\mathcal C(0)t+1}
=\sum_{r=0}^{q-1}\gamma_rm^{\mathcal C(0)t-r}
+\sum_{r=0}^{q-1}\boldsymbol1^T
\left[P^r-\sum_{a=0}^{r-1}\gamma_aP^{r-1-a}\right]
Q\boldsymbol f^{eq}(m^{\mathcal C(0)t-r}).
}
$$

The inner sum is empty for $r=0$. Equilibrium populations remain as explicitly known functions, not additional unknown fields. In particular, no assumption $m^{(\theta)}=m^{eq,(\theta)}$ is made for nonconserved Hermite moments.



### 8. Express gamma_r with your ordered shift subsets

For $I_{r-1}=\{i_0,\ldots,i_{r-1}\}$ with $i_0<\cdots<i_{r-1}$, let $A[I_{r-1},I_{r-1}]$ denote the principal submatrix and set $c_{I_{r-1}}=\sum_{a=0}^{r-1}c_{i_a}$. Then

$$
\boxed{
\gamma_{r-1}=(-1)^{r+1}
\sum_{0\le i_0<\cdots<i_{r-1}\le q-1}
\det A[I_{r-1},I_{r-1}]\,
T_{\Delta x}^{c_{I_{r-1}}},\qquad 1\le r\le q.
}
$$

Indeed, $P[I,I]=\operatorname{diag}(T^{c_i})_{i\in I}A[I,I]$, so its determinant is $T^{\sum_{i\in I}c_i}\det A[I,I]$. The characteristic coefficients are the alternating sums of principal minors.

Consequently,

$$
\begin{aligned}
\gamma_0&=\sum_iA_{ii}T^{c_i},\\
\gamma_1&=-\sum_{i<k}
\begin{vmatrix}A_{ii}&A_{ik}\\A_{ki}&A_{kk}\end{vmatrix}T^{c_i+c_k},\\
\gamma_2&=\sum_{i<k<j}
\begin{vmatrix}A_{ii}&A_{ik}&A_{ij}\\A_{ki}&A_{kk}&A_{kj}\\A_{ji}&A_{jk}&A_{jj}\end{vmatrix}T^{c_i+c_k+c_j}.
\end{aligned}
$$

At the final level,

$$
\gamma_{q-1}=(-1)^{q+1}\left[\prod_{\theta=0}^{q-1}(1-s_\theta)\right]
T^{\sum_{i=0}^{q-1}c_i}.
$$

On a symmetric D1Qq lattice, $\sum_i c_i=0$, so this last shift is the identity. The principal minors encode the MRT coupling that replaces the simple SRT factors $(1-\omega)^r$. For $S=\omega I$, $A=(1-\omega)I$ and the formula reduces to the alternating elementary shift sums of the SRT scheme.



### 9. Fully explicit D1Q3 specialization

Let $(c_0,c_1,c_2)=(0,1,-1)$, $c_s^2=1/3$, $(w_0,w_1,w_2)=(2/3,1/6,1/6)$, and

$$
M=\begin{pmatrix}1&1&1\\0&1&-1\\-1/3&2/3&2/3\end{pmatrix},
\qquad a_\theta=1-s_\theta.
$$

Then

$$
A=\begin{pmatrix}
(2a_0+a_2)/3&2(a_0-a_2)/3&2(a_0-a_2)/3\\
(a_0-a_2)/6&(a_0+2a_2+3a_1)/6&(a_0+2a_2-3a_1)/6\\
(a_0-a_2)/6&(a_0+2a_2-3a_1)/6&(a_0+2a_2+3a_1)/6
\end{pmatrix}.
$$

Writing $T_+=T^{1}$ and $T_-=T^{-1}$,

$$
\begin{aligned}
\gamma_0&=\frac{2a_0+a_2}{3}
+\frac{a_0+2a_2+3a_1}{6}(T_++T_-),\\
\gamma_1&=-\frac{3a_0a_2+2a_0a_1+a_1a_2}{6}(T_++T_-)
-\frac{a_1(a_0+2a_2)}{3},\\
\gamma_2&=a_0a_1a_2.
\end{aligned}
$$

The closed scheme is

$$
\boxed{
\begin{aligned}
m^{\mathcal C(0)t+1}={}&\gamma_0m^{\mathcal C(0)t}
+\gamma_1m^{\mathcal C(0)t-1}+\gamma_2m^{\mathcal C(0)t-2}\\
&+\boldsymbol1^TQ\boldsymbol f^{eq,t}
+\boldsymbol1^T(P-\gamma_0I)Q\boldsymbol f^{eq,t-1}\\
&+\boldsymbol1^T(P^2-\gamma_0P-\gamma_1I)Q\boldsymbol f^{eq,t-2}.
\end{aligned}}
$$

For example, with only density conserved, one may prescribe $m^{eq,(0)}=\rho$, $m^{eq,(1)}=F(\rho)$ and $m^{eq,(2)}=G(\rho)$ and evaluate $\boldsymbol f^{eq}=M^{-1}(\rho,F(\rho),G(\rho))^T$. This makes every term above an explicit function of density history. The general derivation does not require these particular choices of $F$ and $G$.



### 10. Several conserved moments and the moment-space form

Let $\mathcal K$ denote the set of conserved moment labels and

$$
\boldsymbol u=(m^{\mathcal C(n)})_{n\in\mathcal K},\qquad
\boldsymbol C^{(n)}=(\mathcal C_{c_0}^{(n)},\ldots,\mathcal C_{c_{q-1}}^{(n)}).
$$

Assume collision conserves these rows and $\boldsymbol f^{eq}=\boldsymbol f^{eq}(\boldsymbol u)$. Multiplying the vector elimination identity by any such row gives the closed system

$$
\boxed{
m^{\mathcal C(n)t+1}
=\sum_{r=0}^{q-1}\gamma_rm^{\mathcal C(n)t-r}
+\sum_{r=0}^{q-1}\boldsymbol C^{(n)}B_rQ^f
\boldsymbol f^{eq}(\boldsymbol u^{t-r}),\qquad n\in\mathcal K.
}
$$

This $q$-level construction is sufficient but need not be the shortest possible recurrence. It does not require the conserved polynomials to coincide with individual Hermite rows. Conservation itself must be checked: for diagonal relaxation in a Hermite basis, conserved Hermite moments are matched by equilibrium (or have zero relaxation rate); arbitrary linear combinations require the corresponding collision conservation identity.

To remain entirely in moment space, define $B_r^m=MB_rM^{-1}$ and $\boldsymbol\ell^{(n)}=\boldsymbol C^{(n)}M^{-1}$. The same $\gamma_r$ apply because $P^m$ and $P^f$ are similar. Then

$$
m^{\mathcal C(n)t+1}
=\sum_{r=0}^{q-1}\gamma_rm^{\mathcal C(n)t-r}
+\sum_{r=0}^{q-1}\boldsymbol\ell^{(n)}B_r^mQ^m
\boldsymbol m^{eq}(\boldsymbol u^{t-r}).
$$

If $\mathcal C^{(n)}=\mathcal H^{(n)}$, the row $\boldsymbol\ell^{(n)}$ simply selects moment $n$.



### 11. Initialization, scope, and reference

The recurrence uses the $q$ previous levels $t,t-1,\ldots,t-q+1$. To reproduce an LB solution, initialize these levels consistently: start from the intended population state and perform the first $q-1$ LB steps, retaining the conserved moments. Different kinetic initial states with the same initial conserved fields can generate different startup histories. Arbitrary independently selected histories need not represent an admissible LB initialization. Some rates, equilibria, or symmetries allow a lower-order recurrence; $q$ is a sufficient bound here, not a minimality claim.

If the rates or basis depend on position or time, the constant-coefficient trace/minor construction above cannot be used unchanged: multiplication by variable coefficients generally fails to commute with spatial shifts.

The finite elimination principle is the Cayley-Hamilton approach used by Bellotti, Graille and Massot, *Finite Difference formulation of any lattice Boltzmann scheme*, Numerische Mathematik (2022), DOI: [10.1007/s00211-022-01302-2](https://doi.org/10.1007/s00211-022-01302-2), [arXiv:2201.05354](https://arxiv.org/abs/2201.05354). The derivation here is written explicitly in the notebook's notation and connects its final population residual to the principal-minor construction. The $\gamma_r$ indexing here is deliberately chosen to retain the notebook's $\gamma_0=\operatorname{tr}P$; it differs from characteristic-coefficient indexing in that reference.



### 12. Executable verification

The following code checks the D1Q3 characteristic identity exactly with symbolic relaxation rates, verifies the principal-minor coefficients for D1Q3 and D1Q5, and compares direct LB evolution with the closed recurrence using nonlinear equilibria and nonequilibrium initialization. Both one and two conserved moments are tested. The D1Q5 test uses the actual inverse of its Hermite matrix rather than assuming Gaussian discrete orthogonality.


```python
import itertools
import numpy as np
import sympy as sp

z = sp.Symbol('z', nonzero=True)  # T^{c} is represented by z**c


def hermite_matrix(velocities, cs2=sp.Rational(1, 3)):
    q = len(velocities)
    h = [[sp.Integer(1)] * q]
    if q > 1:
        h.append(list(map(sp.Integer, velocities)))
    for degree in range(1, q - 1):
        h.append([velocities[i] * h[degree][i]
                  - degree * cs2 * h[degree - 1][i] for i in range(q)])
    return sp.Matrix(h)


def elimination_operators(velocities, rates):
    q = len(velocities)
    M = hermite_matrix(velocities)
    A = M.inv() * (sp.eye(q) - sp.diag(*rates)) * M
    T = sp.diag(*[z**int(c) for c in velocities])
    P, Q = T * A, T * (sp.eye(q) - A)
    B = [sp.eye(q)]
    gamma = []
    for r in range(1, q + 1):
        product = (P * B[-1]).applyfunc(sp.expand)
        gamma.append(sp.expand(sp.trace(product) / r))
        B.append((product - gamma[-1] * sp.eye(q)).applyfunc(sp.expand))
    assert B[-1] == sp.zeros(q), 'Cayley-Hamilton residual is nonzero'
    for r in range(1, q + 1):
        minor_coefficient = sum(
            A.extract(I, I).det() * z**sum(velocities[i] for i in I)
            for I in itertools.combinations(range(q), r)
        ) * (-1)**(r + 1)
        assert sp.expand(gamma[r - 1] - minor_coefficient) == 0
    return M, A, P, Q, B, gamma


# Exact D1Q3 check, with arbitrary relaxation rates.
s0, s1, s2 = sp.symbols('s0 s1 s2')
M3, A3, P3, Q3, B3, gamma3 = elimination_operators([0, 1, -1], [s0, s1, s2])
a0, a1, a2 = 1 - s0, 1 - s1, 1 - s2
expected_gamma3 = [
    (2*a0 + a2)/3 + (a0 + 2*a2 + 3*a1)/6 * (z + 1/z),
    -(3*a0*a2 + 2*a0*a1 + a1*a2)/6 * (z + 1/z) - a1*(a0 + 2*a2)/3,
    a0*a1*a2,
]
assert all(sp.expand(g - e) == 0 for g, e in zip(gamma3, expected_gamma3))

# Check the explicit pair extraction against the operator residual.
pair_row = sp.zeros(1, 3)
for i, k in itertools.combinations(range(3), 2):
    d = A3[i, i]*A3[k, k] - A3[i, k]*A3[k, i]
    for ell in range(3):
        b = (A3[i, k] - A3[i, i])*A3[k, ell] + (A3[k, i] - A3[k, k])*A3[i, ell]
        if ell in (i, k):
            assert sp.expand(b + d) == 0
        else:
            pair_row[ell] += (b + d) * z**([0, 1, -1][i] + [0, 1, -1][k])
assert (pair_row - sp.ones(1, 3)*B3[2]).applyfunc(sp.expand) == sp.zeros(1, 3)
print('PASS: symbolic D1Q3 coefficients, pair extraction, and B_q = 0.')


def stencil(expression):
    terms = {}
    for term in sp.Add.make_args(sp.expand(expression)):
        power = int(term.as_powers_dict().get(z, 0))
        coefficient = float(term / z**power)
        terms[power] = terms.get(power, 0.0) + coefficient
    return tuple(terms.items())


def apply_stencil(terms, field):
    # np.roll(field, c)[x] = field[x-c], matching the backward-shift convention.
    result = np.zeros_like(field)
    for shift, coefficient in terms:
        result += coefficient * np.roll(field, shift)
    return result


def numerical_check(velocities, rates, conserved_count):
    M, A, P, Q, B, gamma = elimination_operators(velocities, rates)
    q, nx, steps = len(velocities), 64, 16
    Mn, Minv = np.array(M, dtype=float), np.array(M.inv(), dtype=float)
    An = np.array(A, dtype=float)
    Dn = np.eye(q) - An
    C = Mn[:conserved_count]
    gamma_stencils = [stencil(g) for g in gamma]
    source_stencils = []
    for r in range(q):
        source = (M[:conserved_count, :] * B[r] * Q).applyfunc(sp.expand)
        source_stencils.append([[stencil(source[n, j]) for j in range(q)]
                                for n in range(conserved_count)])

    def equilibrium(u):
        rho = u[0]
        moments = np.zeros((q, nx))
        moments[:conserved_count] = u
        for theta in range(conserved_count, q):
            moments[theta] = 0.003*theta*rho**2 + 0.01*np.tanh(rho)
            if conserved_count > 1:
                moments[theta] += 0.005*u[1]**2
        return Minv @ moments

    x = np.arange(nx)
    initial_u = np.zeros((conserved_count, nx))
    initial_u[0] = 1 + 0.04*np.cos(2*np.pi*x/nx)
    if conserved_count > 1:
        initial_u[1] = 0.01*np.sin(2*np.pi*x/nx)
    rng = np.random.default_rng(42 + q + conserved_count)
    f = equilibrium(initial_u) + 1e-3*rng.normal(size=(q, nx))
    direct_history = [C @ f]
    for _ in range(steps):
        feq = equilibrium(C @ f)
        postcollision = An @ f + Dn @ feq
        assert np.max(np.abs(C @ (postcollision - f))) < 1e-12
        f = np.array([np.roll(postcollision[i], velocities[i]) for i in range(q)])
        direct_history.append(C @ f)

    # Only the first q conserved histories are taken from LBM.
    # All later histories are advanced independently by the closed recurrence.
    closed_history = [u.copy() for u in direct_history[:q]]
    for t in range(q - 1, steps):
        next_u = np.zeros((conserved_count, nx))
        for r in range(q):
            old_u = closed_history[t - r]
            old_feq = equilibrium(old_u)
            for n in range(conserved_count):
                next_u[n] += apply_stencil(gamma_stencils[r], old_u[n])
                for j in range(q):
                    next_u[n] += apply_stencil(source_stencils[r][n][j], old_feq[j])
        closed_history.append(next_u)
    error = float(np.max(np.abs(np.array(closed_history) - np.array(direct_history))))
    assert error < 1e-10, error
    print(f'PASS: D1Q{q}, {conserved_count} conserved moment(s), nonlinear equilibrium; '
          f'maximum independent-evolution error = {error:.3e}')


for velocities, rates in [
    ([0, 1, -1], [0, sp.Rational(6, 5), sp.Rational(7, 5)]),
    ([0, 1, -1, 2, -2], [0, sp.Rational(6, 5), sp.Rational(7, 5),
                          sp.Rational(8, 7), sp.Rational(9, 8)]),
]:
    for conserved_count in (1, 2):
        numerical_check(velocities, rates, conserved_count)
```
