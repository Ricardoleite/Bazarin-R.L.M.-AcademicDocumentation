# Generalized Difference Finite Scheme for Balance Equations derived from Lattice Boltzmann

In the remainder of this section, we develop a generalized finite difference scheme (FDS) interpretation of balance equation within the LBM framework. Initially, we dedvelop the analysis under a single-relaxation time (SRT) framework and subsequently we extendo multi-relaxation times (MRT) framework. In both case are considered a generalized lattice $DdQq$.

Given the lattice Boltzmann (LB) equatioin in a SRT formulation:

$$
f_i( x_{\alpha} + c_{i} \delta t, t+\delta t) = f_i(x_{\alpha}, t)  -\left( \frac{ f_{i} - f_{i}^{eq} }{ \tau } \right) \qquad \rightarrow \qquad f_i^{'} = f_i  - \omega\left( f_{i} - f_{i}^{eq} \right), 
$$(LB-fdm-df-Eq)

where $f^{eq}_{i}$ is the equilibrium distribution function that define the moments conserved also called by equilibrium moments in a expansion of Hermite bases:

$$
f^{eq}_{i}\left(m^{\mathcal{C}(0)},...,m^{\mathcal{C}(n)}\right)=w_{i}\displaystyle\sum_{\theta=0}^{N} \displaystyle\frac{ m^{eq,(\theta)}\left(m^{\mathcal{C}(0)},...,m^{\mathcal{C}(n)}\right)\mathcal{H}^{(\theta)} }{\theta! c_{s}^{2\theta} },
$$(general-df-fdm-feq-Eq)

where $m^{eq,(\theta)}$ is a tensor of order $\theta$ called equilibrium expansion coefficient or Hermite equilibrium moments, $\mathcal{H}^{(\theta)}$ is Hermite polynomial tensor of order $\theta$, and $N$ is the lattice order. The terms $m^{\mathcal{C}(0)}(x_{\alpha}, t)$ up to $m^{\mathcal{C}(n)}(x_{\alpha}, t)$ are examples of a consecutive list of $n$ conservative moments that can be recovered by normal product of distribution funcion with specific polynomes $\mathcal{C}_{c_{i}}^{(\theta)}$ due to its ortogonality with Hermite polynomes. So, as example

$$
\begin{array}{c}
\sum_{i}f_{i}\mathcal{C}_{c_{i}}^{(0)}=\displaystyle\sum_{i}f_{i}^{eq}\mathcal{C}_{c_{i}}^{(0)}=m^{\mathcal{C}(0)}, \qquad \sum_{i}f_{i}\mathcal{C}_{c_{i}}^{(1)}=\displaystyle\sum_{i}f_{i}^{eq}\mathcal{C}_{c_{i}}^{(1)}=m^{\mathcal{C}(1)},\\
\sum_{i}f_{i}\mathcal{C}_{c_{i}}^{(2)}=\displaystyle\sum_{i}f_{i}^{eq}\mathcal{C}_{c_{i}}^{(2)}=m^{\mathcal{C}(2)}, \qquad ... \qquad, \textrm{ and } \qquad \sum_{i}f_{i}\mathcal{C}_{c_{i}}^{(\theta)}=\displaystyle\sum_{i}f_{i}^{eq}\mathcal{C}_{c_{i}}^{(\theta)}=m^{\mathcal{C}(\theta)}. 
\end{array}
$$(general-conserved-moments-fdm-fi-df)

The expansion coefficients for each order are recovered by the product of distribution funcion with the respective Hermite polynomial tensor of order, as given in the equation bellow up to the third-order:

$$
\begin{array}{c}
\displaystyle\sum_{i}f_{i}\mathcal{H}^{(0)}=\displaystyle\sum_{i}f_{i}^{eq}\mathcal{H}^{(0)}=m^{eq,(0)}, \qquad \displaystyle\sum_{i}f_{i}^{eq}\mathcal{H}^{(1)}_{\alpha} =m^{eq,(1)}_{\alpha}, \qquad \displaystyle \sum_{i} f_{i}^{eq} \mathcal{H}^{(2)}_{\alpha\beta} =m^{eq,(2)}_{\alpha\beta},\\
\displaystyle\sum_{i} f_{i}^{eq} \mathcal{H}^{(3)}_{\alpha\beta\gamma} = m^{eq,(3)}_{\alpha\beta\gamma}, \qquad ... \qquad, \textrm{ and } \qquad \displaystyle\sum_{i} f_{i}^{eq} \mathcal{H}^{(q)}_{\alpha\beta\gamma} = m^{eq,(q)}_{\alpha\beta\gamma}.
\end{array}
$$(general-moments-fdm-fi-df)

## FDS of Single-Relaxation Time (SRT) LBM framework

For simplification we redefine the notation for $f_i( x_{\alpha} + e_{i,\alpha} \delta t, t+\delta t)=f_{i,x+c_{i}}^{t+1}$. Reinterpreting the Eq. {eq}`LB-fdm-df-Eq` in pre-streaming moment, we apply a shift in $c_{i}$:

$$
f_{i,x_{\alpha}}^{t+1}  = f_{i,x_{\alpha}-c_{i}}^{t}  - \omega \left( f_{i,x_{\alpha}-c_{i}}^{t} - f_{i,x_{\alpha}-c_{i}}^{eq,t} \right).
$$(pre-fdm-LB-df-Eq)

Multiplying by the $n$-order of polynome of the $\mathcal{C}_{c_{i}}^{(n)}$, that correct recover a conserved moment and summing the Eq. {eq}`pre-fdm-LB-df-Eq` over all lattice direction, we have:

$$
\begin{array}{rl}
\displaystyle\sum_{i}\Big( f_{i,x_{\alpha}}^{t+1}\mathcal{C}_{c_{i}}^{(n)} \Big) = & \displaystyle\sum_{i}\Big( f_{i,x_{\alpha}-c_{i}}^{t}\mathcal{C}_{c_{i}}^{(n)}  - \omega \left( f_{i,x_{\alpha}-c_{i}}^{t}\mathcal{C}_{c_{i}}^{(n)} - f_{i,x_{\alpha}-c_{i}}^{eq,t}\mathcal{C}_{c_{i}}^{(n)} \right) \Big), \\
m_{x_{\alpha}}^{\mathcal{C}(n)t+1} = & \displaystyle\sum_{i}\Big( (1-\omega) f_{i,x_{\alpha}-c_{i}}^{t}\mathcal{C}_{c_{i}}^{(n)}  + \textcolor{blue}{\omega f_{i,x_{\alpha}-c_{i}}^{eq,t}\mathcal{C}_{c_{i}}^{(n)} }\Big),
\end{array}
$$

where the blue color highligth the terms that are dependent of only know moments. Given that we know all the moment described in the equilibrium distribution function, we have

$$
m_{x_{\alpha}}^{\mathcal{C}(n)t+1} = \textcolor{blue}{\sum_{i}\left(\omega f_{i,x_{\alpha}-c_{i}}^{eq,t}\mathcal{C}_{c_{i}}^{(n)} \right)} + \sum_{i}\left( (1-\omega) f_{i,x_{\alpha}-c_{i}}^{t}\mathcal{C}_{c_{i}}^{(n)}  \right),
$$

using the Eq. {eq}`general-conserved-moments-fdm-fi-df` to replace $f_{i,x_{\alpha}-c_{i}}^{t}$ in the above equation:

$$
\begin{array}{rl}
m_{x_{\alpha}}^{\mathcal{C}(n)t+1} = & \textcolor{blue}{\displaystyle\sum_{i}\left(\omega f_{i,x_{\alpha}-c_{i}}^{eq,t}\mathcal{C}_{c_{i}}^{(n)} \right)} + \sum_{i}\left( (1-\omega) f_{i,x_{\alpha}-c_{i}}^{t}\mathcal{C}_{c_{i}}^{(n)}  \right), \\
m_{x_{\alpha}}^{\mathcal{C}(n)t+1} = & \textcolor{blue}{\displaystyle\sum_{i}\left(\omega f_{i,x_{\alpha}-c_{i}}^{eq,t}\mathcal{C}_{c_{i}}^{(n)} + (1-\omega)m_{x_{\alpha}-c_{i}}^{\mathcal{C}(n)t} \right)} - (1-\omega) \sum_{i}\left( \sum_{k\neq i} \left(f_{k,x_{\alpha}-c_{i}}^{t}\mathcal{C}_{c_{i}}^{(n)}  \right) \right).
\end{array}
$$

Repeting the process of shift in $c_{k}$ the unknow distribution function and write it terms of pre collisional function, we have:

$$
\begin{array}{rl}
m_{x_{\alpha}}^{\mathcal{C}(n)t+1} = & \textcolor{blue}{\displaystyle\sum_{i}\left(\omega f_{i,x_{\alpha}-c_{i}}^{eq,t}\mathcal{C}_{c_{i}}^{(n)} + (1-\omega)m_{x_{\alpha}-c_{i}}^{\mathcal{C}(n)t} \right)} \\
&- (1-\omega) \displaystyle\sum_{i}\left( \sum_{k\neq i} \left(\left( (1-\omega)f_{k,x_{\alpha}-c_{i}-c_{k}}^{t-1}  + \omega f_{k,x_{\alpha}-c_{i}-c_{k}}^{eq,t-1} \right)\mathcal{C}_{c_{i}}^{(n)}  \right) \right), \\
m_{x_{\alpha}}^{\mathcal{C}(n)t+1} = & \textcolor{blue}{\displaystyle\sum_{i}\left(\omega f_{i,x_{\alpha}-c_{i}}^{eq,t}\mathcal{C}_{c_{i}}^{(n)} + (1-\omega)m_{x_{\alpha}-c_{i}}^{\mathcal{C}(n)t} \right)} - \textcolor{blue}{(1-\omega) \displaystyle\sum_{i}\left( \sum_{k\neq i} \left( \omega f_{k,x_{\alpha}-c_{i}-c_{k}}^{eq,t-1} \mathcal{C}_{c_{i}}^{(n)}  \right) \right)} \\
&- (1-\omega) \displaystyle\sum_{i}\left( \sum_{k\neq i} \left( (1-\omega)f_{k,x_{\alpha}-c_{i}-c_{k}}^{t-1} \mathcal{C}_{c_{i}}^{(n)}  \right) \right),
\end{array}
$$(SRT-FDM-Exp-Step1)

By expansion of terms under the $\sum_{i} \sum_{k\neq i}$, we observe the presence of permutations in the shifts applied. Replace the index the above interpretation in the Eq. {eq}`SRT-FDM-Exp-Step1`, we have

$$

&- (1-\omega) \displaystyle\sum_{i}\left( \sum_{k= i+1} \left( (1-\omega)\left(f_{i,x_{\alpha}-c_{i}-c_{k}}^{t-1} + f_{k,x_{\alpha}-c_{i}-c_{k}}^{t-1} \right) \mathcal{C}_{c_{i}}^{(n)}  \right) \right),

using the Eq. {eq}`general-conserved-moments-fdm-fi-df` to replace $\left(f_{i,x_{\alpha}-c_{i}-c_{k}}^{t-1} + f_{k,x_{\alpha}-c_{i}-c_{k}}^{t-1} \right)$ in the above equation:

\boxed{ \left(f_{i,x_{\alpha}-c_{i}-c_{k}}^{t-1} + f_{k,x_{\alpha}-c_{i}-c_{k}}^{t-1} \right) \mathcal{C}_{c_{i}}^{(n)} = m_{x_{\alpha}-c_{i}-c_{k}}^{\mathcal{C}(n)t-1} - \sum_{j\neq k\neq i} f_{j,x_{\alpha}-c_{i}-c_{k}}^{t-1} \mathcal{C}_{c_{i}}^{(n)} }
$$

$$
\begin{array}{rl}

&- (1-\omega) \displaystyle\sum_{i}\left( \sum_{k= i+1}  (1-\omega)m_{x_{\alpha}-c_{i}-c_{k}}^{\mathcal{C}(n)t-1}  \right) + (1-\omega)^{2} \displaystyle\sum_{i}\left( \sum_{k= i+1} \left( \sum_{j\neq k\neq i} f_{j,x_{\alpha}-c_{i}-c_{k}}^{t-1} \mathcal{C}_{c_{i}}^{(n)} \right) \right), \\
m_{x_{\alpha}}^{\mathcal{C}(n)t+1} = & \textcolor{blue}{\displaystyle\sum_{i}\left(\omega f_{i,x_{\alpha}-c_{i}}^{eq,t}\mathcal{C}_{c_{i}}^{(n)} + (1-\omega)m_{x_{\alpha}-c_{i}}^{\mathcal{C}(n)t} \right)} \\
&- \textcolor{blue}{(1-\omega) \displaystyle\sum_{i}\left( \sum_{k= i+1} \left( \omega \left(f_{i,x_{\alpha}-c_{i}-c_{k}}^{eq,t-1} + f_{k,x_{\alpha}-c_{i}-c_{k}}^{eq,t-1} \right) \mathcal{C}_{c_{i}}^{(n)}  + (1-\omega)m_{x_{\alpha}-c_{i}-c_{k}}^{\mathcal{C}(n)t-1} \right) \right)} \\
&+ (1-\omega)^{2} \displaystyle\sum_{i}\left( \sum_{k= i+1} \left( \sum_{j\neq k\neq i} f_{j,x_{\alpha}-c_{i}-c_{k}}^{t-1} \mathcal{C}_{c_{i}}^{(n)} \right) \right),
\end{array}
$$

Repeting the same processes applied before to find moments in previous time step: 
- write the unknow distribution function in a pre-collisional term shifted in $c_{j}$:

$$
\boxed{  f_{j,x_{\alpha}-c_{i}-c_{k}}^{t-1} =  (1-\omega)f_{j,x_{\alpha}-c_{i}-c_{k}-c_{j}}^{t-2}  + \omega f_{j,x_{\alpha}-c_{i}-c_{k}-c_{j}}^{eq,t-2} }\quad,
$$

- Split the sum to avoid permutations of shifted terms:

$$
\boxed{ \displaystyle\sum_{i}\left( \sum_{k= i+1} \left( \sum_{j\neq k\neq i} f_{j,x_{\alpha}-c_{i}-c_{k}-c_{j}}^{t-2} \right) \right) = \displaystyle\sum_{i}\left( \sum_{k= i+1} \left( \sum_{j= k+1} \left(f_{i,x_{\alpha}-c_{i}-c_{k}-c_{j}}^{t-2} + f_{k,x_{\alpha}-c_{i}-c_{k}-c_{j}}^{t-2} + f_{j,x_{\alpha}-c_{i}-c_{k}-c_{j}}^{t-2} \right) \right) \right) }\quad,
$$

- The Eq. {eq}`general-conserved-moments-fdm-fi-df` to extract from unknow distribution functions the conserved moment:

$$
\boxed{ \left(f_{i,x_{\alpha}-c_{i}-c_{k}-c_{j}}^{t-2} + f_{k,x_{\alpha}-c_{i}-c_{k}-c_{j}}^{t-2} + f_{j,x_{\alpha}-c_{i}-c_{k}-c_{j}}^{t-2} \right) \mathcal{C}_{c_{i}}^{(n)} = m_{x_{\alpha}-c_{i}-c_{k}-c_{j}}^{\mathcal{C}(n)t-2} - \sum_{l\neq j\neq k\neq i} f_{l,x_{\alpha}-c_{i}-c_{k}-c_{j}}^{t-2} \mathcal{C}_{c_{i}}^{(n)} }\quad.
$$

Considering that this process can be repeated $q-1$ times, and replacing the lattice index $i,k,j,l$ by $i_{0},i_{1},i_{2},i_{3}$, respectively, to create a order $i_{n}$ index, we will have:

$$
\begin{array}{rl}
m_{x_{\alpha}}^{\mathcal{C}(n)t+1} = & \textcolor{blue}{\displaystyle\sum_{i_{0}}\left(\omega f_{i_{0},x_{\alpha}-c_{i_{0}}}^{eq,t}\mathcal{C}_{c_{i}}^{(n)} + (1-\omega)m_{x_{\alpha}-c_{i_{0}}}^{\mathcal{C}(n)t} \right)} \\
&- \textcolor{blue}{(1-\omega) \displaystyle\sum_{i_{0}}\left( \sum_{i_{1}= i_{0}+1}  \left(\omega \left(f_{i_{0},x_{\alpha}-c_{i_{0}}-c_{i_{1}}}^{eq,t-1} + f_{i_{1},x_{\alpha}-c_{i_{0}}-c_{i_{1}}}^{eq,t-1} \right) \mathcal{C}_{c_{i}}^{(n)}  + (1-\omega)m_{x_{\alpha}-c_{i_{0}}-c_{i_{1}}}^{\mathcal{C}(n)t-1} \right) \right)} \\
&+ \textcolor{blue}{(1-\omega)^{2} \displaystyle\sum_{i_{0}}\left( \sum_{i_{1}= i_{0}+1} \left( \sum_{i_{2}= i_{1}+1}  \left( \omega \left(\begin{array}{c}f_{i_{0},x_{\alpha}-c_{i_{0}}-c_{i_{1}}-c_{i_{2}}}^{eq,t-2}\\ + f_{i_{1},x_{\alpha}-c_{i_{0}}-c_{i_{1}}-c_{i_{2}}}^{eq,t-2}\\ + f_{i_{2},x_{\alpha}-c_{i_{0}}-c_{i_{1}}-c_{i_{2}}}^{eq,t-2}\end{array} \right) \mathcal{C}_{c_{i}}^{(n)}  + (1-\omega)m_{x_{\alpha}-c_{i_{0}}-c_{i_{1}}-c_{i_{2}}}^{\mathcal{C}(n)t-2} \right) \right) \right)} \\
& \vdots \\
&+ (-1)^{q-1}(1-\omega)^{q-1} \displaystyle\sum_{i_{0}}\left( \sum_{i_{1}= i_{0}+1} \cdots \left( \sum_{i_{q-1}= i_{q-2}+1} \omega\left(\begin{array}{c}f_{i_{0},x_{\alpha}-c_{i_{0}}\cdots-c_{i_{q-1}}}^{t-q+1)} \\ \vdots \\ + f_{i_{q-1},x_{\alpha}-c_{i_{0}}\cdots-c_{i_{q-1}}}^{t-q+1)} \end{array}\right) \mathcal{C}_{c_{i}}^{(n)} \right)\right) \\
& + (-1)^{q-1}(1-\omega)^{q-1} \displaystyle\sum_{i_{0}}\left( \sum_{i_{1}= i_{0}+1} \cdots \left( \sum_{i_{q-1}= i_{q-2}+1}  (1-\omega)m_{x_{\alpha}-c_{i_{0}}\cdots-c_{i_{q-1}}}^{\mathcal{C}(n)t-q+1}  \right) \right),
\end{array}
$$

notice by the lattice symmetry at the step $q-1$ all shifts will be canceled by its opposite, so the last term can reduced to:

$$
\begin{array}{rl}
m_{x_{\alpha}}^{\mathcal{C}(n)t+1} = & \textcolor{blue}{\displaystyle\sum_{i_{0}}\left(\omega f_{i_{0},x_{\alpha}-c_{i_{0}}}^{eq,t}\mathcal{C}_{c_{i}}^{(n)} + (1-\omega)m_{x_{\alpha}-c_{i_{0}}}^{\mathcal{C}(n)t} \right)} \\
&- \textcolor{blue}{(1-\omega) \displaystyle\sum_{i_{0}}\left( \sum_{i_{1}= i_{0}+1} \left( \omega \left(f_{i_{0},x_{\alpha}-c_{i_{0}}-c_{i_{1}}}^{eq,t-1} + f_{i_{1},x_{\alpha}-c_{i_{0}}-c_{i_{1}}}^{eq,t-1} \right) \mathcal{C}_{c_{i}}^{(n)}  + (1-\omega)m_{x_{\alpha}-c_{i_{0}}-c_{i_{1}}}^{\mathcal{C}(n)t-1}\right)  \right)} \\
&+ \textcolor{blue}{(1-\omega)^{2} \displaystyle\sum_{i_{0}}\left( \sum_{i_{1}= i_{0}+1} \left( \sum_{i_{2}= i_{1}+1}  \left(\omega \left(\begin{array}{c}f_{i_{0},x_{\alpha}-c_{i_{0}}-c_{i_{1}}-c_{i_{2}}}^{eq,t-2}\\ + f_{i_{1},x_{\alpha}-c_{i_{0}}-c_{i_{1}}-c_{i_{2}}}^{eq,t-2}\\ + f_{i_{2},x_{\alpha}-c_{i_{0}}-c_{i_{1}}-c_{i_{2}}}^{eq,t-2}\end{array} \right) \mathcal{C}_{c_{i}}^{(n)}  + (1-\omega)m_{x_{\alpha}-c_{i_{0}}-c_{i_{1}}-c_{i_{2}}}^{\mathcal{C}(n)t-2}\right) \right) \right)} \\
& \vdots \\
&+ \textcolor{blue}{(-1)^{q-1}(1-\omega)^{q-1} m_{x_{\alpha}}^{\mathcal{C}(n)t-q+1}},
\end{array}
$$(srt-generalized-expanded-fdm-form)

where all terms are described in terms of know previous moments.

The Eq. {eq}`srt-generalized-expanded-fdm-form` can mathmatically summarized in the form:

$$
\small
\begin{array}{rl}
m_{x_{\alpha}}^{\mathcal{C}(n)t+1} = & \displaystyle \sum_{q_{0}=1}^{q}  \left( (-1)^{q_{0}-1}(1-\omega)^{q_{0}-1} \sum_{0\le i_0 <i_1 <\cdots <i_{q_{0}-1} <q}  \left( \left(\prod_{q_{1} = 1}^{q_{0}}T_{\Delta x}^{c_{q_{1}-1}} \right) \left(\omega \sum_{q_{1}=1}^{q_{0}} f_{i_{q_{1}-1},x_{\alpha}}^{eq,t-q_{0}+1}\mathcal{C}_{c_{i}}^{(n)} + (1-\omega)m_{x_{\alpha}}^{\mathcal{C}(n)t-q_{0}+1}\right) \right)\right),
\end{array}
$$(srt-generalized-fdm-form)

where $T_{\Delta x}^{c_{q_{1}-1}}$ is shift parameter, i.e.,

$$
T_{\Delta x}^{c_{0}} f_{i_{0},x_{\alpha}}^{eq,t} = f_{i_{0},x_{\alpha}-c_{0}}^{eq,t} \qquad \text{and} \qquad T_{\Delta x}^{c_{0}} m_{x_{\alpha}}^{\mathcal{C}(n)t} = m_{x_{\alpha}-c_{0}}^{\mathcal{C}(n)t},
$$

and

$$
\displaystyle\sum_{0\le i_0 <i_1 <\cdots <i_{q_{0}-1} <q} = \sum_{i_0=0}^{q-1} \left( \sum_{i_1=i_0+1}^{q-1} \left( \sum_{i_2=i_1+1}^{q-1} \left( \cdots \sum_{i_{q-1}=i_{q-2}+1}^{q-1} \qquad \right)\right)\right).
$$

The symbolic representation of the expression can be expanded automatically using the SymPy library in Python, as given in the following code:

## FDS of Multi-Relaxation Time (MRT) LBM framework

Given the lattice Boltzmann (LB) equatioin in a generalized MRT formulation:

$$
f_i( x_{\alpha} + e_{i,\alpha} \delta t, t+\delta t) = f_i(x_{\alpha}, t)  - (\textbf{M}^{-1}\textbf{S}\textbf{M})_{ik}\left( f_{k} - f_{k}^{eq}  \right), 
$$(MRT-LB-conv-df-Eq)

where $\textbf{M}$ is matrix of Hermite moments, $\textbf{S}$ is the diagonal matrix of relaxation times, and $\textbf{M}^{-1}$ is the inverse matrix of $\textbf{M}$. Considering the problem description in a lattice $D1Qq$, we have:

$$
\textbf{M} =\begin{pmatrix} \mathcal{H}^{(0)} \\ \mathcal{H}^{(1)}_{\alpha_{1}} \\ \mathcal{H}^{(2)}_{\alpha_{1}\alpha_{2}} \\ \vdots \\ \mathcal{H}^{(q-1)}_{\alpha_{1}\alpha_{2}\cdots\alpha_{q-1}}\end{pmatrix} ,  \qquad \qquad  \textbf{S} = \begin{pmatrix} s_0 & 0   & 0 & \cdots & 0 \\ 0 & s_1 & 0 & \cdots & 0 \\ 0   & 0   & s_2 & \cdots & 0 \\ \vdots & \vdots & \vdots & \ddots & \vdots \\ 0 & 0 & 0 & \cdots & s_{q-1} \end{pmatrix}.
$$

The equilibrium distribution function is given by Eq. {eq}`general-df-fdm-feq-Eq`, and the distribution function is defined by

$$
f_{i} = w_{i}\displaystyle\sum_{\theta=0}^{N}\frac{m^{(\theta)}\mathcal{H}^{(\theta)}}{\theta!c_{s}^{2\theta}}.
$$

where $m^{(\theta)}$ is the Hermite moments also called by expansion coefficients.

Rewritting the collision term:

$$
(\textbf{M}^{-1}\textbf{S}\textbf{M})_{ik}\left( f_{k} - f_{k}^{eq}  \right) = f_{i}^{S} - f_{i}^{S,eq}, 
$$

by transform the distribution function to moments space relaxing each moment and inverting transformation, we have

$$
f_{i}^{S} = w_{i}\displaystyle\sum_{\theta=0}^{q-1}\frac{m^{(\theta)}\mathcal{H}^{(\theta)}}{\theta!c_{s}^{2\theta}s_{\theta}}=w_{i}\left( s_{0} m^{(0)}\mathcal{H}^{(0)} + \frac{s_{1}m^{(1)}_{\alpha_{1}}\mathcal{H}^{(1)}_{\alpha_{1}}}{c_{s}^{2}} + \cdots + \frac{s_{q-1}m^{(q-1)}_{\alpha_{1}\cdots \alpha_{q-1}}\mathcal{H}^{(q-1)}_{\alpha_{1}\cdots \alpha_{q-1}}}{(q-1)!c_{s}^{2(q-1)}} \right).
$$

considering a one-dimensional lattice. It can be extended for d-dimensional lattices described for full projection of one-dimensional lattices.

For simplification we redefine the notation for $f_i( x_{\alpha} + e_{i,\alpha} \delta t, t+\delta t)=f_{i,x+c_{i}}^{t+1}$ and remove the index tensor order notation of Hermite polynomes and moments. Reinterpreting the Eq. {eq}`LB-fdm-df-Eq` in pre-streaming moment, we apply a shift in $c_{i}$:

$$
\begin{array}{rl}
f_{i,x_{\alpha}}^{t+1} = & f_{i,x_{\alpha}-c_{i}}^{t} - \left( f_{i,x_{\alpha}-c_{i}}^{S,t} - f_{i,x_{\alpha}-c_{i}}^{S,eq,t} \right)
\end{array}
$$

Multiplying by the $n$-order of polynome of the $\mathcal{C}_{c_{i}}^{(n)}$, that correct recover a conserved moment and summing the Eq. {eq}`pre-fdm-LB-df-Eq` over all lattice direction, we have:

$$
\begin{array}{rl}
\displaystyle\sum_{i}\Big( f_{i,x_{\alpha}}^{t+1}\mathcal{C}_{c_{i}}^{(n)} \Big) = & \displaystyle\sum_{i}\left(f_{i,x_{\alpha}-c_{i}}^{t}\mathcal{C}_{c_{i}}^{(n)}\right)  - \displaystyle\sum_{i}\left( f_{i,x_{\alpha}-c_{i}}^{S,t} - f_{i,x_{\alpha}-c_{i}}^{S,eq,t} \right)\mathcal{C}_{c_{i}}^{(n)} \\
m_{x_{\alpha}}^{\mathcal{C}(n)t+1} = & \displaystyle\sum_{i}\left(f_{i,x_{\alpha}-c_{i}}^{t}\mathcal{C}_{c_{i}}^{(n)}\right)  - \displaystyle\sum_{i}\left( w_{i}\left(\begin{array}{c} s_{0}m^{(0)}_{x_{\alpha}-c_{i}}\mathcal{H}^{(0)}_{c_{i}} \\+ \displaystyle\frac{s_{1}m^{(1)}_{x_{\alpha}-c_{i}}\mathcal{H}^{(1)}_{c_{i}}}{c_{s}^{2}} \\ \vdots \\+ \displaystyle\frac{s_{q-1}m^{(q-1)}_{x_{\alpha}-c_{i}}\mathcal{H}^{(q-1)}_{c_{i}}}{(q-1)!c_{s}^{2(q-1)}}\end{array} \right) - f_{i,x_{\alpha}-c_{i}}^{S,eq,t} \right)\mathcal{C}_{c_{i}}^{(n)} \\
m_{x_{\alpha}}^{\mathcal{C}(n)t+1} = & \textcolor{blue}{\displaystyle\sum_{i}\left(f_{i,x_{\alpha}-c_{i}}^{S,eq,t}\mathcal{C}_{c_{i}}^{(n)}\right)} + \displaystyle\sum_{i}\left( w_{i}\left( \begin{array}{c} (1-s_{0})m^{(0)}_{x_{\alpha}-c_{i}}\mathcal{H}^{(0)}_{c_{i}} \\ + \displaystyle\frac{(1-s_{1})m^{(1)}_{x_{\alpha}-c_{i}}\mathcal{H}^{(1)}_{c_{i}}}{c_{s}^{2}} \\ \vdots \\+ \displaystyle\frac{(1-s_{q-1})m^{(q-1)}_{x_{\alpha}-c_{i}}\mathcal{H}^{(q-1)}_{c_{i}}}{(q-1)!c_{s}^{2(q-1)}} \end{array} \right) \right)\mathcal{C}_{c_{i}}^{(n)}\\
m_{x_{\alpha}}^{\mathcal{C}(n)t+1} = & \textcolor{blue}{\displaystyle\sum_{i}\left(f_{i,x_{\alpha}-c_{i}}^{S,eq,t}\mathcal{C}_{c_{i}}^{(n)}\right)} + \displaystyle\sum_{i}\left( w_{i}\left(\begin{array}{c} (1-s_{0})\left(\displaystyle\sum_{k}f_{k,x_{\alpha}-c_{i}}\mathcal{H}^{(0)}_{c_{k}}\right)\mathcal{H}^{(0)}_{c_{i}}\\ + \displaystyle\frac{(1-s_{1})\left(\displaystyle\sum_{k}f_{k,x_{\alpha}-c_{i}}\mathcal{H}^{(1)}_{c_{k}}\right)\mathcal{H}^{(1)}_{c_{i}}}{c_{s}^{2}}\\  \vdots \\ + \frac{(1-s_{q-1})\left(\displaystyle\sum_{k}f_{k,x_{\alpha}-c_{i}}\mathcal{H}^{(q-1)}_{c_{k}}\right)\mathcal{H}^{(q-1)}_{c_{i}}}{(q-1)!c_{s}^{2(q-1)}} \end{array} \right) \right)\mathcal{C}_{c_{i}}^{(n)}
\end{array}
$$
