# Generalized Difference Finite Scheme for Balance Equations derived from Lattice Boltzmann

In the remainder of this section, we develop a generalized finite difference scheme (FDS) interpretation of balance equation within the LBM framework. Initially, we dedvelop the analysis under a single relaxation time (SRT) framework and subsequently we extendo MRT framework. In both case are considered a generalized lattice $DdQq$. 

Given the lattice Boltzmann (LB) equatioin in a SRT formulation:

$$
f_i( x_{\alpha} + c_{i} \delta t, t+\delta t) = f_i(x_{\alpha}, t)  -\left( \frac{ f_{i} - f_{i}^{eq} }{ \tau } \right) \qquad \rightarrow \qquad f_i^{'} = f_i  - \omega\left( f_{i} - f_{i}^{eq} \right), 
$$(LB-fdm-df-Eq)

where $f^{eq}_{i}$ is the equilibrium distribution function that define the moments conserved also called by equilibrium moments in a expansion of Hermite bases:

$$
f^{eq}_{i}\left(m^{(0)},...,m^{(n)}\right)=w_{i}\displaystyle\sum_{\theta=0}^{\infty} \displaystyle\frac{ a^{(\theta)}\left(m^{(0)},...,m^{(n)}\right)\mathcal{H}^{(\theta)} }{\theta! c_{s}^{2\theta} },
$$(general-df-fdm-feq-Eq)

where $a^{(\theta)}$ is a tensor of order $\theta$ called expansion coefficient, $\mathcal{H}^{(\theta)}$ is Hermite polynomial tensor of order $\theta$. The terms $m^{(0)}(x_{\alpha}, t)$ up to $m^{(n)}(x_{\alpha}, t)$ are examples of a consecutive list of $n$ conservative moments that can be recovered by normal product of distribution funcion with specific polynomes $\mathcal{C}^{(\theta)}$ due to its ortogonality with Hermite polynomes. So, as example

$$
\begin{array}{c}
\sum_{i}f_{i}\mathcal{C}^{(0)}=\displaystyle\sum_{i}f_{i}^{eq}\mathcal{C}^{(0)}=m^{(0)}, \qquad \sum_{i}f_{i}\mathcal{C}^{(1)}=\displaystyle\sum_{i}f_{i}^{eq}\mathcal{C}^{(1)}=m^{(1)},\\
\sum_{i}f_{i}\mathcal{C}^{(2)}=\displaystyle\sum_{i}f_{i}^{eq}\mathcal{C}^{(2)}=m^{(2)}, \qquad ... \qquad, \textrm{ and } \qquad \sum_{i}f_{i}\mathcal{C}^{(\theta)}=\displaystyle\sum_{i}f_{i}^{eq}\mathcal{C}^{(\theta)}=m^{(\theta)}. 
\end{array}
$$(general-conserved-moments-fdm-fi-df)

The expansion coefficients for each order are recovered by the product of distribution funcion with the respective Hermite polynomial tensor of order, as given in the equation bellow up to the third-order:

$$
\begin{array}{c}
\displaystyle\sum_{i}f_{i}\mathcal{H}^{(0)}=\displaystyle\sum_{i}f_{i}^{eq}\mathcal{H}^{(0)}=a^{(0)}, \qquad \displaystyle\sum_{i}f_{i}^{eq}\mathcal{H}^{(1)}_{\alpha} =a^{(1)}_{\alpha}, \qquad \displaystyle \sum_{i} f_{i}^{eq} \mathcal{H}^{(2)}_{\alpha\beta} =a^{(2)}_{\alpha\beta},\\
\displaystyle\sum_{i} f_{i}^{eq} \mathcal{H}^{(3)}_{\alpha\beta\gamma} = a^{(3)}_{\alpha\beta\gamma}, \qquad ... \qquad, \textrm{ and } \qquad \displaystyle\sum_{i} f_{i}^{eq} \mathcal{H}^{(q)}_{\alpha\beta\gamma} = a^{(q)}_{\alpha\beta\gamma}.
\end{array}
$$(general-moments-fdm-fi-df)

## FDS of Single Relaxation Time (SRT) LBM framework

For simplification we redefine the notation for $f_i( x_{\alpha} + e_{i,\alpha} \delta t, t+\delta t)=f_{i,x+c_{i}}^{t+1}$. Reinterpreting the Eq. {eq}`LB-fdm-df-Eq` in pre-streaming moment, we apply a shift in $c_{i}$:

$$
f_{i,x_{\alpha}}^{t+1}  = f_{i,x_{\alpha}-c_{i}}^{t}  - \omega \left( f_{i,x_{\alpha}-c_{i}}^{t} - f_{i,x_{\alpha}-c_{i}}^{eq,t} \right).
$$(pre-fdm-LB-df-Eq)

Multiplying by the $n$-order of polynome of the $\mathcal{C}^{(n)}$, that correct recover a conserved moment and summing the Eq. {eq}`pre-fdm-LB-df-Eq` over all lattice direction, we have:

$$
\begin{array}{rl}
\displaystyle\sum_{i}\Big( f_{i,x_{\alpha}}^{t+1}\mathcal{C}^{(n)} \Big) = & \displaystyle\sum_{i}\Big( f_{i,x_{\alpha}-c_{i}}^{t}\mathcal{C}^{(n)}  - \omega \left( f_{i,x_{\alpha}-c_{i}}^{t}\mathcal{C}^{(n)} - f_{i,x_{\alpha}-c_{i}}^{eq,t}\mathcal{C}^{(n)} \right) \Big), \\
m_{x_{\alpha}}^{(n)t+1} = & \displaystyle\sum_{i}\Big( (1-\omega) f_{i,x_{\alpha}-c_{i}}^{t}\mathcal{C}^{(n)}  + \textcolor{blue}{\omega f_{i,x_{\alpha}-c_{i}}^{eq,t}\mathcal{C}^{(n)} }\Big),
\end{array}
$$

where the blue color highligth the terms that are dependent of only know moments. Given that we know all the moment described in the equilibrium distribution function, we have

$$
m_{x_{\alpha}}^{(n)t+1} = \textcolor{blue}{\sum_{i}\left(\omega f_{i,x_{\alpha}-c_{i}}^{eq,t}\mathcal{C}^{(n)} \right)} + \sum_{i}\left( (1-\omega) f_{i,x_{\alpha}-c_{i}}^{t}\mathcal{C}^{(n)}  \right),
$$

using the Eq. {eq}`general-conserved-moments-fdm-fi-df` to replace $f_{i,x_{\alpha}-c_{i}}^{t}$ in the above equation:

$$
\begin{array}{rl}
m_{x_{\alpha}}^{(n)t+1} = & \textcolor{blue}{\displaystyle\sum_{i}\left(\omega f_{i,x_{\alpha}-c_{i}}^{eq,t}\mathcal{C}^{(n)} \right)} + \sum_{i}\left( (1-\omega) f_{i,x_{\alpha}-c_{i}}^{t}\mathcal{C}^{(n)}  \right), \\
m_{x_{\alpha}}^{(n)t+1} = & \textcolor{blue}{\displaystyle\sum_{i}\left(\omega f_{i,x_{\alpha}-c_{i}}^{eq,t}\mathcal{C}^{(n)} + (1-\omega)m_{x_{\alpha}-c_{i}}^{(n)t} \right)} - (1-\omega) \sum_{i}\left( \sum_{k\neq i} \left(f_{k,x_{\alpha}-c_{i}}^{t}\mathcal{C}^{(n)}  \right) \right).
\end{array}
$$

Repeting the process of shift in $c_{k}$ the unknow distribution function and write it terms of pre collisional function, we have:

$$
\begin{array}{rl}
m_{x_{\alpha}}^{(n)t+1} = & \textcolor{blue}{\displaystyle\sum_{i}\left(\omega f_{i,x_{\alpha}-c_{i}}^{eq,t}\mathcal{C}^{(n)} + (1-\omega)m_{x_{\alpha}-c_{i}}^{(n)t} \right)} \\
&- (1-\omega) \displaystyle\sum_{i}\left( \sum_{k\neq i} \left(\left( (1-\omega)f_{k,x_{\alpha}-c_{i}-c_{k}}^{t-1}  + \omega f_{k,x_{\alpha}-c_{i}-c_{k}}^{eq,t-1} \right)\mathcal{C}^{(n)}  \right) \right), \\
m_{x_{\alpha}}^{(n)t+1} = & \textcolor{blue}{\displaystyle\sum_{i}\left(\omega f_{i,x_{\alpha}-c_{i}}^{eq,t}\mathcal{C}^{(n)} + (1-\omega)m_{x_{\alpha}-c_{i}}^{(n)t} \right)} - \textcolor{blue}{(1-\omega) \displaystyle\sum_{i}\left( \sum_{k\neq i} \left( \omega f_{k,x_{\alpha}-c_{i}-c_{k}}^{eq,t-1} \mathcal{C}^{(n)}  \right) \right)} \\
&- (1-\omega) \displaystyle\sum_{i}\left( \sum_{k\neq i} \left( (1-\omega)f_{k,x_{\alpha}-c_{i}-c_{k}}^{t-1} \mathcal{C}^{(n)}  \right) \right),
\end{array}
$$(SRT-FDM-Exp-Step1)

Notice that in diagonal from left to right the terms has the same shift with index $i$ and $k$ inverted. In this way, we can rewrite the terms avoiding these shift permutations:

$$
\displaystyle\sum_{i}\left( \sum_{k\neq i} \left( (1-\omega)f_{k,x_{\alpha}-c_{i}-c_{k}}^{t-1} \mathcal{C}^{(n)}  \right) \right)= \displaystyle\sum_{i}\left( \sum_{k= i+1} \left( (1-\omega) \left( f_{i,x_{\alpha}-c_{i}-c_{k}}^{t-1} + f_{k,x_{\alpha}-c_{i}-c_{k}}^{t-1} \right) \mathcal{C}^{(n)}  \right) \right).
$$

Replace the index the above interpretation in the Eq. {eq}`SRT-FDM-Exp-Step1`, we have

$$
\begin{array}{ll}
m_{x_{\alpha}}^{(n)t+1} = & \textcolor{blue}{\displaystyle\sum_{i}\left(\omega f_{i,x_{\alpha}-c_{i}}^{eq,t}\mathcal{C}^{(n)} + (1-\omega)m_{x_{\alpha}-c_{i}}^{(n)t} \right)} - \textcolor{blue}{(1-\omega) \displaystyle\sum_{i}\left( \sum_{k= i+1} \left( \omega \left(f_{i,x_{\alpha}-c_{i}-c_{k}}^{eq,t-1} + f_{k,x_{\alpha}-c_{i}-c_{k}}^{eq,t-1} \right) \mathcal{C}^{(n)}  \right) \right)} \\
&- (1-\omega) \displaystyle\sum_{i}\left( \sum_{k= i+1} \left( (1-\omega)\left(f_{i,x_{\alpha}-c_{i}-c_{k}}^{t-1} + f_{k,x_{\alpha}-c_{i}-c_{k}}^{t-1} \right) \mathcal{C}^{(n)}  \right) \right),
\end{array}
$$

using the Eq. {eq}`general-conserved-moments-fdm-fi-df` to replace $\left(f_{i,x_{\alpha}-c_{i}-c_{k}}^{t-1} + f_{k,x_{\alpha}-c_{i}-c_{k}}^{t-1} \right)$ in the above equation:

$$
\boxed{ \left(f_{i,x_{\alpha}-c_{i}-c_{k}}^{t-1} + f_{k,x_{\alpha}-c_{i}-c_{k}}^{t-1} \right) \mathcal{C}^{(n)} = m_{x_{\alpha}-c_{i}-c_{k}}^{(n)t-1} - \sum_{j\neq k\neq i} f_{j,x_{\alpha}-c_{i}-c_{k}}^{t-1} \mathcal{C}^{(n)} }
$$

$$
\begin{array}{rl}
m_{x_{\alpha}}^{(n)t+1} = & \textcolor{blue}{\displaystyle\sum_{i}\left(\omega f_{i,x_{\alpha}-c_{i}}^{eq,t}\mathcal{C}^{(n)} + (1-\omega)m_{x_{\alpha}-c_{i}}^{(n)t} \right)} - \textcolor{blue}{(1-\omega) \displaystyle\sum_{i}\left( \sum_{k= i+1} \left( \omega \left(f_{i,x_{\alpha}-c_{i}-c_{k}}^{eq,t-1} + f_{k,x_{\alpha}-c_{i}-c_{k}}^{eq,t-1} \right) \mathcal{C}^{(n)}  \right) \right)} \\
&- (1-\omega) \displaystyle\sum_{i}\left( \sum_{k= i+1}  (1-\omega)m_{x_{\alpha}-c_{i}-c_{k}}^{(n)t-1}  \right) + (1-\omega)^{2} \displaystyle\sum_{i}\left( \sum_{k= i+1} \left( \sum_{j\neq k\neq i} f_{j,x_{\alpha}-c_{i}-c_{k}}^{t-1} \mathcal{C}^{(n)} \right) \right), \\
m_{x_{\alpha}}^{(n)t+1} = & \textcolor{blue}{\displaystyle\sum_{i}\left(\omega f_{i,x_{\alpha}-c_{i}}^{eq,t}\mathcal{C}^{(n)} + (1-\omega)m_{x_{\alpha}-c_{i}}^{(n)t} \right)} \\
&- \textcolor{blue}{(1-\omega) \displaystyle\sum_{i}\left( \sum_{k= i+1} \left( \omega \left(f_{i,x_{\alpha}-c_{i}-c_{k}}^{eq,t-1} + f_{k,x_{\alpha}-c_{i}-c_{k}}^{eq,t-1} \right) \mathcal{C}^{(n)}  + (1-\omega)m_{x_{\alpha}-c_{i}-c_{k}}^{(n)t-1} \right) \right)} \\
&+ (1-\omega)^{2} \displaystyle\sum_{i}\left( \sum_{k= i+1} \left( \sum_{j\neq k\neq i} f_{j,x_{\alpha}-c_{i}-c_{k}}^{t-1} \mathcal{C}^{(n)} \right) \right),
\end{array}
$$

Repeting the same processes applied before to find moments in previous time step: 
- write the unknow distribution function in a pre-collisional term shifted in $c_{j}$:

$$
\boxed{  f_{j,x_{\alpha}-c_{i}-c_{k}}^{t-1} =  (1-\omega)f_{j,x_{\alpha}-c_{i}-c_{k}-c_{j}}^{t-2}  + \omega f_{j,x_{\alpha}-c_{i}-c_{k}-c_{j}}^{eq,t-2} }\quad,
$$

- Split the sum to avoid permutations of shifted terms:

$$
\boxed{ \displaystyle\sum_{i}\left( \sum_{k= i+1} \left( \sum_{j\neq k\neq i} f_{j,x_{\alpha}-c_{i}-c_{k}-c_{j}}^{t-2} \right) \right) = \displaystyle\sum_{i}\left( \sum_{k= i+1} \left( \sum_{j= k+1} \left(f_{i,x_{\alpha}-c_{i}-c_{k}-c_{j}}^{t-2} + f_{k,x_{\alpha}-c_{i}-c_{k}-c_{j}}^{t-2} + f_{j,x_{\alpha}-c_{i}-c_{k}-c_{j}}^{t-2} \right) \right) \right) }\quad,
$$

- The Eq. {eq}`general-conserved-moments-fdm-fi-df` to extract from unknow distribution functions the conserved moment:

$$
\boxed{ \left(f_{i,x_{\alpha}-c_{i}-c_{k}-c_{j}}^{t-2} + f_{k,x_{\alpha}-c_{i}-c_{k}-c_{j}}^{t-2} + f_{j,x_{\alpha}-c_{i}-c_{k}-c_{j}}^{t-2} \right) \mathcal{C}^{(n)} = m_{x_{\alpha}-c_{i}-c_{k}-c_{j}}^{(n)t-2} - \sum_{l\neq j\neq k\neq i} f_{l,x_{\alpha}-c_{i}-c_{k}-c_{j}}^{t-2} \mathcal{C}^{(n)} }\quad.
$$

Considering that this process can be repeated $q-1$ times, and replacing the lattice index $i,k,j,l$ by $i_{0},i_{1},i_{2},i_{3}$, respectively, to create a order $i_{n}$ index, we will have:

$$
\begin{array}{rl}
m_{x_{\alpha}}^{(n)t+1} = & \textcolor{blue}{\displaystyle\sum_{i_{0}}\left(\omega f_{i_{0},x_{\alpha}-c_{i_{0}}}^{eq,t}\mathcal{C}^{(n)} + (1-\omega)m_{x_{\alpha}-c_{i_{0}}}^{(n)t} \right)} \\
&- \textcolor{blue}{(1-\omega) \displaystyle\sum_{i_{0}}\left( \sum_{i_{1}= i_{0}+1}  \omega \left(f_{i_{0},x_{\alpha}-c_{i_{0}}-c_{i_{1}}}^{eq,t-1} + f_{i_{1},x_{\alpha}-c_{i_{0}}-c_{i_{1}}}^{eq,t-1} \right) \mathcal{C}^{(n)}  + (1-\omega)m_{x_{\alpha}-c_{i_{0}}-c_{i_{1}}}^{(n)t-1}  \right)} \\

& \vdots \\
&+ (-1)^{q-1}(1-\omega)^{q-1} \displaystyle\sum_{i_{0}}\left( \sum_{i_{1}= i_{0}+1} \cdots \left( \sum_{i_{q-1}= i_{q-2}+1} \left(f_{i_{0},x_{\alpha}-c_{i_{0}}\cdots-c_{i_{q-1}}}^{t-q+1)} + \cdots + f_{i_{q-1},x_{\alpha}-c_{i_{0}}\cdots-c_{i_{q-1}}}^{t-q+1)}\right) \mathcal{C}^{(n)} \right) + (1-\omega)m_{x_{\alpha}-c_{i_{0}}\cdots-c_{i_{q-1}}}^{(n)t-q+1}  \right),
\end{array}
$$

$$
\begin{array}{rl}
m_{x_{\alpha}}^{(n)t+1} = & \textcolor{blue}{\displaystyle\sum_{i_{0}}\left(\omega f_{i_{0},x_{\alpha}-c_{i_{0}}}^{eq,t}\mathcal{C}^{(n)} + (1-\omega)m_{x_{\alpha}-c_{i_{0}}}^{(n)t} \right)} \\
&- \textcolor{blue}{(1-\omega) \displaystyle\sum_{i_{0}}\left( \sum_{i_{1}= i_{0}+1}  \omega \left(f_{i_{0},x_{\alpha}-c_{i_{0}}-c_{i_{1}}}^{eq,t-1} + f_{i_{1},x_{\alpha}-c_{i_{0}}-c_{i_{1}}}^{eq,t-1} \right) \mathcal{C}^{(n)}  + (1-\omega)m_{x_{\alpha}-c_{i_{0}}-c_{i_{1}}}^{(n)t-1}  \right)} \\
&+ \textcolor{blue}{(1-\omega)^{2} \displaystyle\sum_{i_{0}}\left( \sum_{i_{1}= i_{0}+1} \left( \sum_{i_{2}= i_{1}+1}  \omega \left(\begin{array}{c}f_{i_{0},x_{\alpha}-c_{i_{0}}-c_{i_{1}}-c_{i_{2}}}^{eq,t-2}\\ + f_{i_{1},x_{\alpha}-c_{i_{0}}-c_{i_{1}}-c_{i_{2}}}^{eq,t-2}\\ + f_{i_{2},x_{\alpha}-c_{i_{0}}-c_{i_{1}}-c_{i_{2}}}^{eq,t-2}\end{array} \right) \mathcal{C}^{(n)}  + (1-\omega)m_{x_{\alpha}-c_{i_{0}}-c_{i_{1}}-c_{i_{2}}}^{(n)t-2} \right) \right)} \\
& \vdots \\
&+ (-1)^{q-1}(1-\omega)^{q-1} \displaystyle\sum_{i_{0}}\left( \sum_{i_{1}= i_{0}+1} \cdots \left( \sum_{i_{q-1}= i_{q-2}+1} \left(\begin{array}{c}f_{i_{0},x_{\alpha}-c_{i_{0}}\cdots-c_{i_{q-1}}}^{t-q+1)} \\ \vdots \\ + f_{i_{q-1},x_{\alpha}-c_{i_{0}}\cdots-c_{i_{q-1}}}^{t-q+1)} \end{array}\right) \mathcal{C}^{(n)} \right) + (1-\omega)m_{x_{\alpha}-c_{i_{0}}\cdots-c_{i_{q-1}}}^{(n)t-q+1}  \right),

\begin{array}{rl}

where all terms are described in terms of know previous moments.
